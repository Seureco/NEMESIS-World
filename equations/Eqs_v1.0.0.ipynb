# Calculating missing data

```python
io.identities["TIME"] = "2015"
io.identities.execute("TIME", f"{SMPSTRY}", f"{SMPSTRY}") 
io.identities["TIME"] = "TIME[-1]+1"
io.identities.execute("TIME", "2016Y1", f"{SMPENDY}") 
```

```python
## Values by (r,s) for DMAT, DINV, DFINEN, PROD, ADDDENRTOT, ADDDMAT, ADDCONSH, ADDCONSNP, ADDCONSG, ADDDINV, ADDDSTOCK
## Values by (r,s) for CONSG, CONSNP or (r,cp) for CONSH 
## Values and equations by (r,s) for DENRTOTQ
## Values for RRLR by (r) 

# Volume equales values in 2019 for DMAT, DINV, DFINEN, PROD, ADDDENRTOT, ADDDMAT, ADDCONSH, ADDCONSNP, ADDCONSG, ADDDINV, ADDDSTOCK
lst_var_vol_sec = ["DMAT", "DINV", "DFINEN", "PROD", "ADDDENRTOT", "ADDDMAT", "ADDCONSH", "ADDCONSNP", "ADDCONSG", "ADDDINV", "ADDDSTOCK"]
for r in lst_reg:
    for s in lst_sec_dfinen_firms:
        for var in lst_var_vol_sec:
            io.identities[f"{var}Q_{r}_{s}"] = f"{var}V_{r}_{s}"
            io.identities.execute(f"{var}Q_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")
            if var == "DFINEN":
                for enrp in lst_enrp_dfinen:
                    io.identities[f"{var}Q_{r}_{s}_{enrp}"] = f"{var}V_{r}_{s}_{enrp}"
                    io.identities.execute(f"{var}Q_{r}_{s}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")
    for s, lst_enrp_tmp in dict_lst_sec_dinten.items():
        io.identities[f"DINTENQ_{r}_{s}"] = f"DINTENV_{r}_{s}"
        io.identities.execute(f"DINTENQ_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")
        for enrp in lst_enrp_tmp:
            io.identities[f"DINTENQ_{r}_{s}_{enrp}"] = f"DINTENV_{r}_{s}_{enrp}"
            io.identities.execute(f"DINTENQ_{r}_{s}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")
    for s, lst_enrp_tmp in dict_lst_sec_dnonen.items():
        io.identities[f"DNONENQ_{r}_{s}"] = f"DNONENV_{r}_{s}"
        io.identities.execute(f"DNONENQ_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")
        for enrp in lst_enrp_tmp:
            io.identities[f"DNONENQ_{r}_{s}_{enrp}"] = f"DNONENV_{r}_{s}_{enrp}"
            io.identities.execute(f"DNONENQ_{r}_{s}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")

# Volume equales values in 2019 for CONS*
lst_var_vol_reg = ["CONSH", "CONSG", "CONSNP"]


for r in lst_reg:
    for var in lst_var_vol_reg:
        io.identities[f"{var}Q_{r}"] = f"{var}V_{r}"
        io.identities.execute(f"{var}Q_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    for cs in lst_bundle_conso_all_lvl_enr:
        io.identities[f"CONSHQ_{r}_{cs}"] = f"CONSHV_{r}_{cs}"
        io.identities.execute(f"CONSHQ_{r}_{cs}", f"{SMPSTRY}", f"{SMPENDY}")   


# Calcul of DENRTOTQ (sum DFINENQ, DINTENQ and DNONENQ if applicable)
for reg in lst_reg:
    for s in lst_sec:
        exp = []
        if s in lst_sec_dfinen_firms:
            exp.append(f"DFINENQ_{reg}_{s}")
        if s in lst_sec_dinten:
            exp.append(f"DINTENQ_{reg}_{s}")
        if s in lst_sec_dnonen:
            exp.append(f"DNONENQ_{reg}_{s}")
        io.equations[f"DENRTOTQ_{reg}_{s}"] = f"DENRTOTQ_{reg}_{s} := " + "+".join(exp)
        io.identities[f"DENRTOTQ_{reg}_{s}"] = "+".join(exp)
        io.identities.execute(f"DENRTOTQ_{reg}_{s}", f"{SMPSTRY}", f"{SMPENDY}")

# RRLR
for r in lst_reg:
    io.identities[f"RRLR_{r}"] = f"RZLR_{r}-INFL_{r}"
    io.identities.execute(f"RRLR_{r}", f"{SMPSTRY}", f"{SMPENDY}")
```

```python
## Values by (r,s) for PINV, PPRODHT, PMAT, PFINEN, PINTEN, PNONEN, PENRTOT, PIMP

# Prices set to one
lst_prices_idt = ["PINV", "PPRODHT", "PMAT", "PFINEN", "PINTEN", "PNONEN", "PENRTOT", "PIMP", "PEXP"]
for r in lst_reg:
    for s in lst_sec:
        for prices in lst_prices_idt:
            io.identities[f"{prices}_{r}_{s}"] = "1"
            io.identities.execute(f"{prices}_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")
```

```python
## Values by (r,skill,s) and (r,skill) for PEMPHOUR

# Caution : There are some unconsistency between Employment data and Compensation of employees, a correction is done to avoid dummy values (several cases where the compensation of employees is positive while employemnt is nil and conversely)
# PEMPHOUR
for r in lst_reg:
    for skill in lst_Lab_skill:
        io.identities[f"PEMPHOUR{skill}_{r}"] = f"CMPEMPA{skill}_{r}/DEMPHOUR{skill}_{r}"
        io.identities.execute(f"PEMPHOUR{skill}_{r}", f"{SMPSTRY}", f"{SMPENDY}")
        for s in lst_sec:
            # io.identities[f"PEMPHOUR{skill}_{r}_{s}"] = f"if(DEMPHOUR{skill}_{r}_{s}<0.01 or CMPEMPA{skill}_{r}_{s}<0.001,PEMPHOUR{skill}_{r},CMPEMPA{skill}_{r}_{s}/DEMPHOUR{skill}_{r}_{s})"
            io.identities[f"PEMPHOUR{skill}_{r}_{s}"] = f"if(DEMPHOUR{skill}_{r}_{s}[{BASEYEAR}]<0.01 or CMPEMPA{skill}_{r}_{s}[{BASEYEAR}]<0.01,PEMPHOUR{skill}_{r},CMPEMPA{skill}_{r}_{s}/DEMPHOUR{skill}_{r}_{s})"
            io.identities.execute(f"PEMPHOUR{skill}_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")
```

```python
## Values by (r,s) for PIMP, PADDDEM, PRODQ, IMPQ, PPROD
## Equations by (r,s) for PADDDEM

# Other variables
for r in lst_reg:
    for sec in lst_sec:
        io.identities[f"PIMP_{r}_{sec}"] = "1"
        io.identities.execute(f"PIMP_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
        io.identities[f"PADDDEM_{r}_{sec}"] = "1"
        io.identities.execute(f"PADDDEM_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
        io.identities[f"PRODQ_{r}_{sec}"] = f"PRODV_{r}_{sec}"
        io.identities.execute(f"PRODQ_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
        io.identities[f"IMPQ_{r}_{sec}"] = f"IMPV_{r}_{sec}"
        io.identities.execute(f"IMPQ_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
        io.identities[f"PPROD_{r}_{sec}"] = f"1"
        io.identities.execute(f"PPROD_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
# And equations
        io.equations[f"PADDDEM_{r}_{sec}"] = f"PADDDEM_{r}_{sec} := (PRODQ_{r}_{sec}*PPROD_{r}_{sec}+IMPQ_{r}_{sec}*PIMP_{r}_{sec})/(PRODQ_{r}_{sec}+IMPQ_{r}_{sec})"
```

```python
## Values by (r) for PCONSH, CONSHQ, INCRDISP

# Other requested variables
# Prices set to one
lst_prices_idt = ["PCONSH"]
for r in lst_reg:
        for prices in lst_prices_idt:
            io.identities[f"{prices}_{r}"] = "1"
            io.identities.execute(f"{prices}_{r}", f"{SMPSTRY}", f"{SMPENDY}")

for r in lst_reg:
    io.identities[f"CONSHQ_{r}"] = f"CONSHV_{r}"
    io.identities[f"INCRDISPH_{r}"] = f"INCGDISPH_{r}"
    io.identities.execute(f"CONSHQ_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.identities.execute(f"INCRDISPH_{r}", f"{SMPSTRY}", f"{SMPENDY}")            
```

```python
## Values by (r,cs) for PCONSH, CONSHQ, CONSHAL
## Values and equations by (r) for PCONSH

# Aggregated consumption price
for r in lst_reg:
    # for cs in lst_bundle_conso_all_lvl:
    for cs in lst_bundle_conso_all_lvl_enr:
        io.identities[f"CONSHQ_{r}_{cs}"] = f"CONSHV_{r}_{cs}"
        io.identities[f"CONSHAL_{r}_{cs}"] = f"CONSHQ_{r}_{cs}"
        io.identities[f"PCONSH_{r}_{cs}"] = "1"
        io.identities.execute(f"CONSHQ_{r}_{cs}", f"{SMPSTRY}", f"{SMPENDY}")
        io.identities.execute(f"CONSHAL_{r}_{cs}", f"{SMPSTRY}", f"{SMPENDY}")
        io.identities.execute(f"PCONSH_{r}_{cs}", f"{SMPSTRY}", f"{SMPENDY}")

for r in lst_reg:
    # terms_num = [f"PCONSH_{r}_{cs}*CONSHQ_{r}_{cs}" for cs in lst_bundle_conso_last_lvl]
    # terms_den = [f"PCONSH_{r}_{cs}[{BASEYEAR}]*CONSHQ_{r}_{cs}" for cs in lst_bundle_conso_last_lvl]
    terms_num = [f"PCONSH_{r}_{cs}*CONSHQ_{r}_{cs}" for cs in lst_bundle_conso_last_lvl_enr]
    terms_den = [f"PCONSH_{r}_{cs}[{BASEYEAR}]*CONSHQ_{r}_{cs}" for cs in lst_bundle_conso_last_lvl_enr]
    idt = "(" + "+".join(terms_num) + ")/(" + "+".join(terms_den) + ")"
    exp = f"PCONSH_{r} := (" + "+".join(terms_num) + ")/(" + "+".join(terms_den) + ")"
    io.identities[f"PCONSH_{r}"] = idt
    io.identities.execute(f"PCONSH_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.equations[f"PCONSH_{r}"] = exp
```

```python
## Values by (r,s) for IMPQ/EXPV/EXPQ

# Calculation of variables: IMPQ_rd_sd_rs, EXPV_rd_sd, EXPQ_rd_sd
lst_idt_torun = []

for rd in lst_reg:
    for sd in lst_sec:
        for rs in lst_reg:
            if rs != rd:
                # IMPQ = IMPV
                io.identities[f"IMPQ_{rd}_{sd}_{rs}"] = f"IMPV_{rd}_{sd}_{rs}"
                lst_idt_torun.append(f"IMPQ_{rd}_{sd}_{rs}")
        # io.identities[f"IMPQ_{rd}_{sd}"] = "+".join(f"IMPQ_{rd}_{sd}_{rs}" for rs in lst_reg if rs != rd)
        # lst_idt_torun.append(f"IMPQ_{rd}_{sd}")        
        io.identities[f"IMPQ_{rd}_{sd}"] = f"IMPV_{rd}_{sd}"
        lst_idt_torun.append(f"IMPQ_{rd}_{sd}")          
        io.identities[f"EXPV_{rd}_{sd}"] = "+".join(f"IMPV_{rs}_{sd}_{rd}" for rs in lst_reg if rs != rd)
        lst_idt_torun.append(f"EXPV_{rd}_{sd}")
        io.identities[f"EXPQ_{rd}_{sd}"] = f"EXPV_{rd}_{sd}"
        lst_idt_torun.append(f"EXPQ_{rd}_{sd}")
        

for idt in lst_idt_torun:
    io.identities.execute(f"{idt}", f"{SMPSTRY}", f"{SMPENDY}")          
```

```python
## Values by (r,s) for KLE, PKLE, KL, PKL, KHSMS, PKHSMS, KHS, PKHS, DFINENQ, PDFINENQ, DENROTHQ, PENROTH

for r in lst_reg:
    for s in lst_sec:
        bundles = {f"PRODQ_{r}_{s}": [[f"KLE_{r}_{s}",f"DMATQ_{r}_{s}"],[f"PKLE_{r}_{s}",f"PMAT_{r}_{s}"]],
                        f"KLE_{r}_{s}": [[f"KL_{r}_{s}",f"DFINENQ_{r}_{s}"],[f"PKL_{r}_{s}",f"PFINEN_{r}_{s}"]],
                        f"KL_{r}_{s}": [[f"KHSMS_{r}_{s}",f"DEMPHOURLS_{r}_{s}"],[f"PKHSMS_{r}_{s}",f"PEMPHOURLS_{r}_{s}"]],
                        f"KHSMS_{r}_{s}": [[f"KHS_{r}_{s}",f"DEMPHOURMS_{r}_{s}"],[f"PKHS_{r}_{s}",f"PEMPHOURMS_{r}_{s}"]],
                        f"KHS_{r}_{s}": [[f"DINVQ_{r}_{s}",f"DEMPHOURHS_{r}_{s}"],[f"USERKCOST_{r}_{s}",f"PEMPHOURHS_{r}_{s}"]],
                        f"DFINENQ_{r}_{s}": [[f"DENROTHQ_{r}_{s}", f"DFINENQ_{r}_{s}_ELEC"],[f"PENROTH_{r}_{s}", f"PENRTAX_{r}_{s}_ELEC"]],
                        f"DENROTHQ_{r}_{s}": [[f"DFINENQ_{r}_{s}_{enrp}" for enrp in lst_enrp_dfinen if enrp != "ELEC"],[f"PENRTAX_{r}_{s}_{enrp}" for enrp in lst_enrp_dfinen if enrp != "ELEC"]]}

        for bdlup, (bdl_q_lst, bdl_p_lst) in reversed(bundles.items()):
            if not bdlup.startswith("PRODQ"):
                bdlq_up_idt = "+".join(f"{bdl_q}*{bdl_p}" for (bdl_q, bdl_p) in zip(bdl_q_lst, bdl_p_lst))
                io.identities[f"{bdlup}"] = bdlq_up_idt
                if bdlup.startswith("DENROTHQ"):
                    io.identities[f"PENROTH_{r}_{s}"] = "1"
                elif bdlup.startswith("DFINENQ"):
                    io.identities[f"PFINEN"] = "1"
                else:
                    io.identities[f"P{bdlup}"] = "1"
                    

# Identities execution (order is important) 
for r in lst_reg:
    for s in lst_sec:
        bundles = {f"PRODQ_{r}_{s}": [[f"KLE_{r}_{s}",f"DMATQ_{r}_{s}"],[f"PKLE_{r}_{s}",f"PMAT_{r}_{s}"]],
                f"KLE_{r}_{s}": [[f"KL_{r}_{s}",f"DFINENQ_{r}_{s}"],[f"PKL_{r}_{s}",f"PFINEN_{r}_{s}"]],
                f"KL_{r}_{s}": [[f"KHSMS_{r}_{s}",f"DEMPHOURLS_{r}_{s}"],[f"PKHSMS_{r}_{s}",f"PEMPHOURLS_{r}_{s}"]],
                f"KHSMS_{r}_{s}": [[f"KHS_{r}_{s}",f"DEMPHOURMS_{r}_{s}"],[f"PKHS_{r}_{s}",f"PEMPHOURMS_{r}_{s}"]],
                f"KHS_{r}_{s}": [[f"DINVQ_{r}_{s}",f"DEMPHOURHS_{r}_{s}"],[f"USERKCOST_{r}_{s}",f"PEMPHOURHS_{r}_{s}"]],
                f"DFINENQ_{r}_{s}": [[f"DENROTHQ_{r}_{s}", f"DFINENQ_{r}_{s}_ELEC"],[f"PENROTH_{r}_{s}", f"PENRTAX_{r}_{s}_ELEC"]],
                f"DENROTHQ_{r}_{s}": [[f"DFINENQ_{r}_{s}_{enrp}" for enrp in lst_enrp_dfinen if enrp != "ELEC"],[f"PENRTAX_{r}_{s}_{enrp}" for enrp in lst_enrp_dfinen if enrp != "ELEC"]]}
        
        demun = []
        denum_tot =  []

        for bdlup, (bdl_q_lst, bdl_p_lst) in reversed(bundles.items()):
            denum = [f"{bdlq}[{BASEYEAR}]*{bdlp}[{BASEYEAR}]" for bdlq, bdlp in zip(bdl_q_lst, bdl_p_lst)]
            denum_tot = "+".join(denum)
            if not bdlup.startswith("DFINENQ"):
                io.identities.execute(f"{bdlup}", f"{SMPSTRY}", f"{SMPENDY}")
            for bdlq, bdlp in zip(bdl_q_lst, bdl_p_lst):
                num = f"({bdlq}[{BASEYEAR}]*{bdlp}[{BASEYEAR}])"
                nb_bdl = len(bdl_q_lst)
                # io.identities[f"S{bdlq}"] = f"if({bdlup}[{BASEYEAR}]<0.01,1/{nb_bdl},{num}/({denum_tot}))"
                io.identities[f"S{bdlq}"] = f"if({bdlup}[{BASEYEAR}]<0.01,1/{nb_bdl},if({bdlq}[{BASEYEAR}]<0.01,0.001,{num}/({denum_tot})))"
                io.identities.execute(f"S{bdlq}", f"{SMPSTRY}", f"{SMPENDY}")
            # io.identities.execute(f"{bdlup}", f"{SMPSTRY}", f"{SMPENDY}")
            if not bdlup.startswith("PRODQ"):
                if bdlup != "PRODQ":
                    if bdlup.startswith("DENROTHQ"):
                        io.identities.execute(f"PENROTH_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")
                    elif bdlup.startswith("DFINENQ"):
                        io.identities.execute(f"PFINEN_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")
                    else:
                        io.identities.execute(f"P{bdlup}", f"{SMPSTRY}", f"{SMPENDY}")
                        
```

```python
## Values by (r,s) for TPPRODQ, TPKLE, TPDMATQ, TPKL, TPDFINENQ, TPKHSMS, TPDEMPHOURLS, TPKHS, TPDEMPHOURMS, TPDINVQ, TPDEMPHOURHS, TPDENROTHQ, TPDEMPHOURHS, 
## Equations by (r,s) for USERKCOST, KLE, DMATQ, KL, DFINENQ, KHSMS, DEMPHOURLS, KHS, DEMPHOURMS, DINVQ, DEMPHOURHS, DENRTOTQ, DFINENQ_ELEC, DFINENQ_{src}
## Equations by (r,s) for PKLE, PKL, PFINEN, PKHSMS, PKHS, PENROTH

for r in lst_reg:
    for s in lst_sec:

        # Define USERKCOST equation 
        io.equations[f"USERKCOST_{r}_{s}"] = f"USERKCOST_{r}_{s} := PINV_{r}_{s}*(RRLR_{r}+decay_{r.lower()}_{s})"
        
        # Define all bundles
        bundles = {f"PRODQ_{r}_{s}": [[f"KLE_{r}_{s}",f"DMATQ_{r}_{s}"],[f"PKLE_{r}_{s}",f"PMAT_{r}_{s}"]],
                f"KLE_{r}_{s}": [[f"KL_{r}_{s}",f"DFINENQ_{r}_{s}"],[f"PKL_{r}_{s}",f"PFINEN_{r}_{s}"]],
                f"KL_{r}_{s}": [[f"KHSMS_{r}_{s}",f"DEMPHOURLS_{r}_{s}"],[f"PKHSMS_{r}_{s}",f"PEMPHOURLS_{r}_{s}"]],
                f"KHSMS_{r}_{s}": [[f"KHS_{r}_{s}",f"DEMPHOURMS_{r}_{s}"],[f"PKHS_{r}_{s}",f"PEMPHOURMS_{r}_{s}"]],
                f"KHS_{r}_{s}": [[f"DINVQ_{r}_{s}",f"DEMPHOURHS_{r}_{s}"],[f"USERKCOST_{r}_{s}",f"PEMPHOURHS_{r}_{s}"]],
                f"DFINENQ_{r}_{s}": [[f"DENROTHQ_{r}_{s}", f"DFINENQ_{r}_{s}_ELEC"],[f"PENROTH_{r}_{s}", f"PENRTAX_{r}_{s}_ELEC"]],
                f"DENROTHQ_{r}_{s}": [[f"DFINENQ_{r}_{s}_{enrp}" for enrp in lst_enrp_dfinen if enrp != "ELEC"],[f"PENRTAX_{r}_{s}_{enrp}" for enrp in lst_enrp_dfinen if enrp != "ELEC"]]}

        # Reverse path to calculate from bottom to top
        for bdlup, (bdl_q_lst, bdl_p_lst) in reversed(bundles.items()):

            # Calculate initial values for TP (Technological progress, set one one)
            io.identities[f"TP{bdlup}"] = "1"
            io.identities.execute(f"TP{bdlup}", f"{SMPSTRY}", f"{SMPENDY}")
            io.equations[f"TP{bdlup}"] = f"TP{bdlup} := if(TIME<2019,1,TP{bdlup}[-1]*exp(-PT{bdlup}))"
            io.identities[f"PT{bdlup}"] = "0"
            io.identities.execute(f"PT{bdlup}", f"{SMPSTRY}", f"{SMPENDY}")
            
            for bdl_q in bdl_q_lst:
                io.identities[f"TP{bdl_q}"] = "1"
                io.identities.execute(f"TP{bdl_q}", f"{SMPSTRY}", f"{SMPENDY}")
                io.equations[f"TP{bdl_q}"] = f"TP{bdl_q} := if(TIME<2019,1,TP{bdl_q}[-1]*exp(-PT{bdl_q}))"
                io.identities[f"PT{bdl_q}"] = "0"
                io.identities.execute(f"PT{bdl_q}", f"{SMPSTRY}", f"{SMPENDY}")

            # CES equations by bundle
            for bdl_q, bdl_p in zip(bdl_q_lst, bdl_p_lst):
                bldup_shrt = bdlup.split("_")[0]
                sigma = f"sigma_{bldup_shrt.lower()}_{r.lower()}_{s}"

                nested_sum = " + ".join(f"S{bdl_q1}[{BASEYEAR}]**({sigma})*({bdl_p1}*TP{bdl_q1})**(1-{sigma})" for bdl_q1, bdl_p1 in zip(bdl_q_lst, bdl_p_lst))

                eq_ces = f"{bdl_q} := if({bdl_q}[{BASEYEAR}]<0.01,{bdl_q}[{BASEYEAR}],C{bdl_q}*({bdlup}*TP{bdlup})" + f"*S{bdl_q}[{BASEYEAR}]**({sigma})*({bdl_p}*TP{bdl_q})**(-{sigma})" + f"*({nested_sum})**({sigma}/(1-{sigma})))"
                io.equations[f"{bdl_q}"] = eq_ces

                # Calibration identities by bundle
                cal_eq = f"{bdl_q}/(({bdlup}*TP{bdlup})" + f"*S{bdl_q}[{BASEYEAR}]**({sigma})*({bdl_p}*TP{bdl_q})**(-{sigma})" + f"*({nested_sum})**({sigma}/(1-{sigma})))"
                io.identities[f"C{bdl_q}"] = cal_eq
                io.identities.execute(f"C{bdl_q}", f"{SMPSTRY}", f"{SMPENDY}")

            # Calculate bundle price (P) including for PRODQ
            if not bdlup.startswith("PRODQ"):
                bdlup_p_tmp = " + ".join(f"(S{bdl_q}[{BASEYEAR}]**({sigma})*({bdl_p}*TP{bdl_q})**(1-{sigma}))" for bdl_q, bdl_p in zip(bdl_q_lst, bdl_p_lst))
                bdlup_p_tmp_penr = " + ".join(f"(S{bdl_q}[{BASEYEAR}]*(({bdl_p}/{bdl_p}[{BASEYEAR}])*TP{bdl_q})**(1-{sigma}))" for bdl_q, bdl_p in zip(bdl_q_lst, bdl_p_lst))
                if bdlup.startswith("DENROTHQ"):
                    io.equations[f"PENROTH_{r}_{s}"] = (f"PENROTH_{r}_{s} := CALPENROTH_{r}_{s}[{BASEYEAR}]*({bdlup_p_tmp_penr})**(1/(1-{sigma}))")
                    io.identities[f"CALPENROTH_{r}_{s}"] = f"PENROTH_{r}_{s}/({bdlup_p_tmp_penr})**(1/(1-{sigma}))"
                    io.identities.execute(f"CALPENROTH_{r}_{s}", SMPSTRY, SMPENDY)
                elif bdlup.startswith("DFINENQ"):
                    io.equations[f"PFINEN_{r}_{s}"] = (f"PFINEN_{r}_{s} := CALPFINEN_{r}_{s}[{BASEYEAR}]*({bdlup_p_tmp_penr})**(1/(1-{sigma}))")
                    io.identities[f"CALPFINEN_{r}_{s}"] = f"PFINEN_{r}_{s}/({bdlup_p_tmp_penr})**(1/(1-{sigma}))"
                    io.identities.execute(f"CALPFINEN_{r}_{s}", SMPSTRY, SMPENDY)
                else:
                    io.equations[f"P{bdlup}"] = (f"P{bdlup} := CALP{bdlup}[{BASEYEAR}]*({bdlup_p_tmp})**(1/(1-{sigma}))")
                    io.identities[f"CALP{bdlup}"] = f"P{bdlup}/({bdlup_p_tmp})**(1/(1-{sigma}))"
                    io.identities.execute(f"CALP{bdlup}", SMPSTRY, SMPENDY)
            else:
                bdlup_p_tmp = f"(SKLE_{r}_{s}[{BASEYEAR}]**({sigma})*(PKLE_{r}_{s}*TPKLE_{r}_{s})**(1-{sigma}))+(SDMATQ_{r}_{s}[{BASEYEAR}]**({sigma})*(PMAT_{r}_{s}*TPDMATQ_{r}_{s})**(1-{sigma}))"
                io.equations[f"PPRODHT_{r}_{s}"] = f"PPRODHT_{r}_{s} := CALPPRODHT_{r}_{s}[{BASEYEAR}]*({bdlup_p_tmp})**(1/(1-{sigma}))"
                io.identities[f"CALPPRODHT_{r}_{s}"] = f"PPRODHT_{r}_{s}/(({bdlup_p_tmp})**(1/(1-{sigma})))"
                io.identities.execute(f"CALPPRODHT_{r}_{s}", SMPSTRY, SMPENDY)
                  
```

```python
### Values and equations by (r,s,enrp) for DENRTOTQ
### Equations by (r,s) for ADDDENRTOTQ

## Addressed demand for energy with energy module
# Caution: Not applied to "SBM" & "IW" as not initally accounted as 
dict_map_sup_enrp = {"01": ["SBM"], "03": ["SBM"], "05": ["COMB"], "06": ["OIL"], "07": ["GAS"], "13": ["COMB"], "14": ["OIL","LBF"], 
                     "34": ["IW"], "35": ["ELEC"], "36": ["ELEC"], "37": ["ELEC"], "38": ["GAS","BG"], "39": ["HEAT"]}

# Invese the dictionnary 
dict_map_sup_enrp_inv = defaultdict(list)
for k, vals in dict_map_sup_enrp.items():
    for v in vals:
        dict_map_sup_enrp_inv[v].append(k)


# Only keep values in several keys
dict_map_sup_enrp_duplic = {v: keys for v, keys in dict_map_sup_enrp_inv.items() if len(keys) > 1}       

# Calculate DENRTOTQ by energy product
for reg in lst_reg:
    for s in lst_sec:
        for enrp in lst_enrp_dfinen:
            exp = []
            if s in lst_sec_dfinen_firms :
                exp.append(f"DFINENQ_{reg}_{s}_{enrp}")
            if s in lst_sec_dinten and enrp in dict_lst_sec_dinten[s]:
                exp.append(f"DINTENQ_{reg}_{s}_{enrp}")
            if s in lst_sec_dnonen and enrp in dict_lst_sec_dnonen[s]:
                exp.append(f"DNONENQ_{reg}_{s}_{enrp}")
            io.equations[f"DENRTOTQ_{reg}_{s}_{enrp}"] = f"DENRTOTQ_{reg}_{s}_{enrp} := " + "+".join(exp)
            io.identities[f"DENRTOTQ_{reg}_{s}_{enrp}"] = "+".join(exp)
            io.identities.execute(f"DENRTOTQ_{reg}_{s}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")

for r in lst_reg:
    for ss, lst_enrp_tmp in dict_map_sup_enrp.items():
        eq_terms = []
        for enrp in lst_enrp_tmp:
            if ss in lst_enr_sect:
                for sd in lst_sec:
                    denum_terms = []
                    if enrp in dict_map_sup_enrp_duplic.keys():
                        denum_terms = [f"menr_{r.lower()}_{sd}_{sx}" for sx in dict_map_sup_enrp_duplic[enrp] if sx in lst_enr_sect]
                        denum = "+".join(denum_terms)
                        eq_terms.append(f"(menr_{r.lower()}_{sd}_{ss}/({denum}))*DENRTOTQ_{r}_{sd}_{enrp}")
                    else:
                        eq_terms.append(f"DENRTOTQ_{r}_{sd}_{enrp}")
                exp = "+".join(eq_terms)
                io.identities[f"ADDDENRTOTQTMP_{r}_{ss}"] = exp
                io.identities.execute(f"ADDDENRTOTQTMP_{r}_{ss}", f"{SMPSTRY}", f"{SMPENDY}")
                io.identities[f"CALADDDENRTOTQ_{r}_{ss}"] = f"ADDDENRTOTQ_{r}_{ss}/ADDDENRTOTQTMP_{r}_{ss}"
                io.identities.execute(f"CALADDDENRTOTQ_{r}_{ss}", f"{SMPSTRY}", f"{SMPENDY}")
                io.equations[f"ADDDENRTOTQ_{r}_{ss}"] = f"ADDDENRTOTQ_{r}_{ss} := if(ADDDENRTOTQTMP_{r}_{ss}<0.001,ADDDENRTOTQ_{r}_{ss}[{BASEYEAR}],CALADDDENRTOTQ_{r}_{ss}[{BASEYEAR}]*({exp}))"
            else:
                io.identities[f"ADDDENRTOTQTMP_{r}_{ss}"] = "0"
                io.identities.execute(f"ADDDENRTOTQTMP_{r}_{ss}", f"{SMPSTRY}", f"{SMPENDY}")
                io.equations[f"ADDDENRTOTQ_{r}_{ss}"] = f"ADDDENRTOTQ_{r}_{ss} := 0"
```

```python
## Equations by (r,s,enrp) for PENRHT

for r in lst_reg:
    for sd in lst_sec:
        for enrp, lst_sec_enr_temp in dict_map_sup_enrp_inv.items():
            if enrp != "SBM":
                eq_terms = []
                if len(lst_sec_enr_temp) >1:
                    denum_terms = [f"menr_{r.lower()}_{sd}_{senr}" for senr in dict_map_sup_enrp_inv[enrp]]
                    denum = "+".join(denum_terms)
                    eq_terms = [f"(menr_{r.lower()}_{sd}_{senr}/({denum}))*(PADDDEM_{r}_{senr}/PADDDEM_{r}_{senr}[{BASEYEAR}])" for senr in lst_sec_enr_temp]
                    exp = "+".join(eq_terms)
                    io.equations[f"PENRHT_{r}_{sd}_{enrp}"] = f"PENRHT_{r}_{sd}_{enrp} := CALPENRHT_{r}_{sd}_{enrp}*({exp})"
                    io.identities[f"CALPENRHT_{r}_{sd}_{enrp}"] = f"PENRHT_{r}_{sd}_{enrp}/({exp})"
                    io.identities.execute(f"CALPENRHT_{r}_{sd}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")
                else:
                    for senr in lst_sec_enr_temp:
                        io.equations[f"PENRHT_{r}_{sd}_{enrp}"] = f"PENRHT_{r}_{sd}_{enrp} := CALPENRHT_{r}_{sd}_{enrp}*(PADDDEM_{r}_{senr}/PADDDEM_{r}_{senr}[{BASEYEAR}])"
                        io.identities[f"CALPENRHT_{r}_{sd}_{enrp}"] = f"PENRHT_{r}_{sd}_{enrp}/(PADDDEM_{r}_{senr}/PADDDEM_{r}_{senr}[{BASEYEAR}])"
                        io.identities.execute(f"CALPENRHT_{r}_{sd}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")
            else:
                io.equations[f"PENRHT_{r}_{sd}_{enrp}"] = f"PENRHT_{r}_{sd}_{enrp} := CALPENRHT_{r}_{sd}_{enrp}*((1/3)*PADDDEM_{r}_01/PADDDEM_{r}_01[{BASEYEAR}]+(2/3)*PADDDEM_{r}_03/PADDDEM_{r}_03[{BASEYEAR}])"
                io.identities[f"CALPENRHT_{r}_{sd}_{enrp}"] = f"PENRHT_{r}_{sd}_{enrp}/((1/3)*PADDDEM_{r}_01/PADDDEM_{r}_01[{BASEYEAR}]+(2/3)*PADDDEM_{r}_03/PADDDEM_{r}_03[{BASEYEAR}])"
                io.identities.execute(f"CALPENRHT_{r}_{sd}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")
    # For GEO
    io.equations[f"PENRHT_{r}_{sd}_GEO"] = f"PENRHT_{r}_{sd}_GEO := CALPENRHT_{r}_{sd}_GEO*((1/3)*PADDDEM_{r}_25/PADDDEM_{r}_25[{BASEYEAR}]+(2/3)*PADDDEM_{r}_41/PADDDEM_{r}_41[{BASEYEAR}])"
    io.identities[f"CALPENRHT_{r}_{sd}_GEO"] = f"PENRHT_{r}_{sd}_GEO/((1/3)*PADDDEM_{r}_25/PADDDEM_{r}_25[{BASEYEAR}]+(2/3)*PADDDEM_{r}_41/PADDDEM_{r}_41[{BASEYEAR}])"
    io.identities.execute(f"CALPENRHT_{r}_{sd}_GEO", f"{SMPSTRY}", f"{SMPENDY}")
    # For SLT
    io.equations[f"PENRHT_{r}_{sd}_SOLT"] = f"PENRHT_{r}_{sd}_SOLT := CALPENRHT_{r}_{sd}_SOLT*((1/3)*PADDDEM_{r}_41/PADDDEM_{r}_41[{BASEYEAR}]+(2/3)*PADDDEM_{r}_28/PADDDEM_{r}_28[{BASEYEAR}])"
    io.identities[f"CALPENRHT_{r}_{sd}_SOLT"] = f"PENRHT_{r}_{sd}_SOLT/((1/3)*PADDDEM_{r}_41/PADDDEM_{r}_41[{BASEYEAR}]+(2/3)*PADDDEM_{r}_28/PADDDEM_{r}_28[{BASEYEAR}])"
    io.identities.execute(f"CALPENRHT_{r}_{sd}_SOLT", f"{SMPSTRY}", f"{SMPENDY}")
```

```python
## Equations by (r,cs) for CONSHQ, PCONSH, PCONSHHT (last level)
## Equations by (r,cs) PCONSHHT (with cs last levels)

### Equations CES (quantities and prices)
for r in lst_reg:
    # for node, child in dict_bundle_conso.items():
    for node, child in dict_bundle_conso_enr.items():
        to_sum = [f"SCONSHQ_{r}_{ch}**(sigma_cons_{node.lower()})*PCONSH_{r}_{ch}**(1-sigma_cons_{node.lower()})" for ch in child]
        exp = "+".join(to_sum)
        for ch in child:
            if node == "TOT":
                io.equations[f"CONSHAL_{r}_{ch}"] = f"CONSHAL_{r}_{ch} := if(CONSHAL_{r}_{ch}[{BASEYEAR}]<0.001,CONSHAL_{r}_{ch}[{BASEYEAR}],CALCONSHAL_{r}_{ch}*CONSHQ_{r}*SCONSHQ_{r}_{ch}**(sigma_cons_{node.lower()})*(PCONSH_{r}_{node}/PCONSH_{r}_{ch})**(sigma_cons_{node.lower()}))"
                io.identities[f"CALCONSHAL_{r}_{ch}"] = f"CONSHAL_{r}_{ch}/(CONSHQ_{r}*SCONSHQ_{r}_{ch}**(sigma_cons_{node.lower()})*(PCONSH_{r}_{node}/PCONSH_{r}_{ch})**(sigma_cons_{node.lower()}))"
                io.identities.execute(f"CALCONSHAL_{r}_{ch}", f"{SMPSTRY}", f"{SMPENDY}")
            else:
                io.equations[f"CONSHAL_{r}_{ch}"] = f"CONSHAL_{r}_{ch} := if(CONSHAL_{r}_{ch}[{BASEYEAR}]<0.001,CONSHAL_{r}_{ch}[{BASEYEAR}],CALCONSHAL_{r}_{ch}*CONSHAL_{r}_{node}*SCONSHQ_{r}_{ch}**(sigma_cons_{node.lower()})*(PCONSH_{r}_{node}/PCONSH_{r}_{ch})**(sigma_cons_{node.lower()}))"
                io.identities[f"CALCONSHAL_{r}_{ch}"] = f"CONSHAL_{r}_{ch}/(CONSHAL_{r}_{node}*SCONSHQ_{r}_{ch}**(sigma_cons_{node.lower()})*(PCONSH_{r}_{node}/PCONSH_{r}_{ch})**(sigma_cons_{node.lower()}))"
                io.identities.execute(f"CALCONSHAL_{r}_{ch}", f"{SMPSTRY}", f"{SMPENDY}")
        io.equations[f"PCONSH_{r}_{node}"] = f"PCONSH_{r}_{node} := CALPCONSH_{r}_{node}*({exp})**(1/(1-sigma_cons_{node.lower()}))"
        io.identities[f"CALPCONSH_{r}_{node}"] = f"PCONSH_{r}_{node}/(({exp})**(1/(1-sigma_cons_{node.lower()})))"
        io.identities.execute(f"CALPCONSH_{r}_{node}", f"{SMPSTRY}", f"{SMPENDY}")


### Equations prices (last level) to connect with PADDDEM
# Inverse dict_map_conso
inverse_map_conso = {}
inverse_map_conso_enr = {}

# for agg, secs in dict_map_conso.items():
for agg, secs in dict_map_conso_enr.items():
    if isinstance(secs, str):  # valeur simple
        secs = [secs]
    for sec in secs:
        # inverse_map_conso.setdefault(sec, []).append(agg)
        inverse_map_conso_enr.setdefault(sec, []).append(agg)


for r in lst_reg:
    # for key, items in inverse_map_conso.items():
    for key, items in inverse_map_conso_enr.items():
        if len(items) > 1:
            demun = [f"mconsh_{r.lower()}_{key.lower()}_{cs}" for cs in items]
            demun_tot = "+".join(demun)
            terms = [f"(mconsh_{r.lower()}_{key.lower()}_{cs}/({demun_tot}))*PADDDEM_{r}_{cs}" for cs in items]
            terms_tax = [f"(mconsh_{r.lower()}_{key.lower()}_{cs}/({demun_tot}))*vatrate_{r.lower()}_{cs.lower()}" for cs in items]
        else:
            terms = [f"PADDDEM_{r}_{cs}" for cs in items]
            terms_tax = [f"vatrate_{r.lower()}_{cs.lower()}" for cs in items]
        exp = f"PCONSHTAX_{r}_{key} := " + "+".join(terms)
        io.equations[f"PCONSHTAX_{r}_{key}"] = exp
        io.identities[f"PCONSHTAX_{r}_{key}"] = "+".join(terms)
        io.identities.execute(f"PCONSHTAX_{r}_{key}", f"{SMPSTRY}", f"{SMPENDY}")
        io.identities[f"RVATAX_{r}_{key}"] = "+".join(terms_tax)
        io.identities.execute(f"RVATAX_{r}_{key}", f"{SMPSTRY}", f"{SMPENDY}")
        # io.equations[f"PCONSH_{r}_{key}"] = f"PCONSH_{r}_{key} := CALPCONSH_{r}_{key}[{BASEYEAR}]*PCONSHTAX_{r}_{key}*(1+vatrate_{r.lower()}_{key.lower()})"
        io.equations[f"PCONSH_{r}_{key}"] = f"PCONSH_{r}_{key} := CALPCONSH_{r}_{key}[{BASEYEAR}]*PCONSHTAX_{r}_{key}*(1+RVATAX_{r}_{key})"
        io.identities[f"CALPCONSH_{r}_{key}"] = f"PCONSH_{r}_{key}/(PCONSHTAX_{r}_{key}*(1+RVATAX_{r}_{key}))"
        io.identities.execute(f"CALPCONSH_{r}_{key}", f"{SMPSTRY}", f"{SMPENDY}")

lst_consp_trade = ["ESSE", "WAWA", "HTCL", "PRVC", "EQUP"]

## TRADE consumption purpose (fixed share of products)
for r in lst_reg:
    exp = [f"CONSHAL_{r}_{trd}" for trd in lst_consp_trade]
    sum = "+".join(exp)
    io.equations[f"CONSHAL_{r}_TRAD"] = f"CONSHAL_{r}_TRAD := CALCONSHAL_{r}_TRAD[{BASEYEAR}]*({sum})"
    io.identities[f"CALCONSHAL_{r}_TRAD"] = f"CONSHAL_{r}_TRAD/({sum})"
    io.identities.execute(f"CALCONSHAL_{r}_TRAD", f"{SMPSTRY}", f"{SMPENDY}")


# Sum all CONSHAL for the last level of the bundle
for r in lst_reg:
    sum = [f"CONSHAL_{r}_{nd}" for nd in lst_bundle_conso_last_lvl_enr]
    io.equations[f"CONSHAL_{r}_TOT"] = f"CONSHAL_{r}_TOT := " + "+".join(sum)


# Re-scale CONSHQ to match CONSHQ agregated
for r in lst_reg:
    for nd in lst_bundle_conso_all_lvl_enr:
        io.equations[f"CONSHQ_{r}_{nd}"] = f"CONSHQ_{r}_{nd} := CONSHAL_{r}_{nd}*(CONSHQ_{r}/CONSHAL_{r}_TOT)"
        
```

```python
## Equations by (r,cs,enrp) for cs in HC and TR for PENRHT, PENRTAX, DFINENP
## Equations and values by (r,cs) for cs in HC and TR for PCONSHTAX
## Values by (r,cs,enrp) for cs in HC and TR for TAXEM

# PENRHT/PENRTTAX HC and TR
for r in lst_reg:
    for sec_conso, dict_tmp in {"HC": dict_conso_enr_hc, "TR": dict_cons_enrp_sec_map_tr}.items():
        for cs, sub_dict_tmp in dict_tmp.items():
            for enrp, sec_lst_tmp in sub_dict_tmp.items():
                # PENRHT
                if len(sec_lst_tmp) > 1:
                    demun = [f"mconsh_{r.lower()}_{cs.lower()}_{sec}" for sec in sec_lst_tmp]
                    demun_tot = "+".join(demun)
                    terms = [f"(mconsh_{r.lower()}_{cs.lower()}_{sec}/({demun_tot}))*PADDDEM_{r}_{sec}/PADDDEM_{r}_{sec}[{BASEYEAR}]" for sec in sec_lst_tmp]
                else:
                    terms = [f"PADDDEM_{r}_{sec}/PADDDEM_{r}_{sec}[{BASEYEAR}]" for sec in sec_lst_tmp]
                sum = "+".join(terms)
                io.equations[f"PENRHT_{r}_{sec_conso}_{enrp}"] = f"PENRHT_{r}_{sec_conso}_{enrp} := CALPENRHT_{r}_{sec_conso}_{enrp}[{BASEYEAR}]*({sum})"
                io.identities[f"CALPENRHT_{r}_{sec_conso}_{enrp}"] = f"PENRHT_{r}_{sec_conso}_{enrp}/({sum})"
                io.identities.execute(f"CALPENRHT_{r}_{sec_conso}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")
                # PENRTAX
                io.variables[f"TAXEM_{r}_{sec_conso}_{enrp}"] = 0
                io.equations[f"PENRTAX_{r}_{sec_conso}_{enrp}"] = f"PENRTAX_{r}_{sec_conso}_{enrp} := PENRHT_{r}_{sec_conso}_{enrp}+TAXENR_{r}_{sec_conso}_{enrp}+TAXEM_{r}_{sec_conso}_{enrp}"
                io.identities[f"PENRTAX_{r}_{sec_conso}_{enrp}"] = f"PENRHT_{r}_{sec_conso}_{enrp}+TAXENR_{r}_{sec_conso}_{enrp}+TAXEM_{r}_{sec_conso}_{enrp}"
                io.identities.execute(f"PENRTAX_{r}_{sec_conso}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")
                # PCONSHHT
                io.equations[f"PCONSHTAX_{r}_{cs}"] = f"PCONSHTAX_{r}_{cs} := PENRTAX_{r}_{sec_conso}_{enrp}/PENRTAX_{r}_{sec_conso}_{enrp}[{BASEYEAR}]"
                io.identities[f"PCONSHTAX_{r}_{cs}"] = f"PENRTAX_{r}_{sec_conso}_{enrp}/PENRTAX_{r}_{sec_conso}_{enrp}[{BASEYEAR}]"
                io.identities.execute(f"PCONSHTAX_{r}_{cs}", f"{SMPSTRY}", f"{SMPENDY}")

       
for sec_conso, dict_tmp in {"HC": dict_conso_enr_hc, "TR": dict_cons_enrp_sec_map_tr}.items():
    for r in lst_reg:
        for cs, dict_enrp in dict_tmp.items():
            for enrp in dict_enrp.keys():
                io.equations[f"DFINENP_{r}_{sec_conso}_{enrp}"] = f"DFINENP_{r}_{sec_conso}_{enrp} := if(DFINENP_{r}_{sec_conso}_{enrp}[{BASEYEAR}]<0.001,DFINENP_{r}_{sec_conso}_{enrp}[{BASEYEAR}],CDFINENP_{r}_{sec_conso}_{enrp}[{BASEYEAR}]*(CONSHQ_{r}_{cs}/CONSHQ_{r}_{cs}[{BASEYEAR}]))"
                io.identities[f"CDFINENP_{r}_{sec_conso}_{enrp}"] = f"DFINENP_{r}_{sec_conso}_{enrp}/(CONSHQ_{r}_{cs}/CONSHQ_{r}_{cs}[{BASEYEAR}])"
                io.identities.execute(f"CDFINENP_{r}_{sec_conso}_{enrp}", f"{SMPSTRY}", f"{SMPENDY}")       
```

## Trade

```python
## Values by (rd,sd,rs) for TAXIMP
## Values and equations by (rd,sd,rs) for PIMP
## Equations by (rd,sd,rs) for IMPQ
## Equations by (rs,ss) for PEXP, EXPQ, PIMP
## Values and equations by (rs,ss) for PEXPCOMP

# Trade block
for rd in lst_reg:
    for sd in lst_sec:
        for rs in lst_reg:
            if rs != rd:
                io.equations[f"PIMP_{rd}_{sd}_{rs}"] = f"PIMP_{rd}_{sd}_{rs} := PEXP_{rs}_{sd}*TAXIMP_{rd}_{sd}_{rs}"
                io.identities[f"TAXIMP_{rd}_{sd}_{rs}"] = "1"
                io.identities[f"PIMP_{rd}_{sd}_{rs}"] = f"PEXP_{rs}_{sd}*TAXIMP_{rd}_{sd}_{rs}[{BASEYEAR}]"

        io.equations[f"PEXP_{rd}_{sd}"] = f"dln(PEXP_{rd}_{sd}) := CALPEXP_{rd}_{sd}+cexp_pint_{rd.lower()}_{sd}*dln(PEXPCOMP_{rd}_{sd})+cexp_pdom_{rd.lower()}_{sd}*dln(PPRODHT_{rd}_{sd})+cexp_exp_{rd.lower()}_{sd}*(ln(PEXP_{rd}_{sd}[-1])-(lexp_pdom_{rd.lower()}_{sd}*ln(PPRODHT_{rd}_{sd}[-1])))"
        exp = "+".join(f"PPRODHT_{rs}_{sd}*EXPQ_{rs}_{sd}" for rs in lst_reg if rs != rd)
        exptot = "+".join(f"PPRODHT_{rs}_{sd}[{BASEYEAR}]*EXPQ_{rs}_{sd}" for rs in lst_reg if rs != rd)
        io.equations[f"PEXPCOMP_{rd}_{sd}"] = f"PEXPCOMP_{rd}_{sd} := " + f"({exp})/({exptot})"
        io.identities[f"PEXPCOMP_{rd}_{sd}"] = f"({exp})/({exptot})"
        io.identities[f"CALPEXP_{rd}_{sd}"] = f"dln(PEXP_{rd}_{sd})-(cexp_pint_{rd.lower()}_{sd}*dln(PEXPCOMP_{rd}_{sd})+cexp_pdom_{rd.lower()}_{sd}*dln(PPRODHT_{rd}_{sd})+cexp_exp_{rd.lower()}_{sd}*(ln(PEXP_{rd}_{sd}[-1])-(lexp_pdom_{rd.lower()}_{sd}*ln(PPRODHT_{rd}_{sd}[-1]))))"
        
# EXPQ_rs_sd        
for rs in lst_reg:
    for sd in lst_sec:        
        io.equations[f"EXPQ_{rs}_{sd}"] = f"EXPQ_{rs}_{sd} := " + "+".join(f"IMPQ_{rd}_{sd}_{rs}" for rd in lst_reg if rd != rs)   


for rd in lst_reg:
    for sd in lst_sec:
        # Calculation of imports share by origin
        terms = [f"IMPQ_{rd}_{sd}_{rs}*PIMP_{rd}_{sd}_{rs}" for rs in lst_reg if rs != rd]
        exp = "+".join(terms)
        for rs in lst_reg:
            if rd != rs:
                io.identities[f"SHIMPQ_{rd}_{sd}_{rs}"] = f"(IMPQ_{rd}_{sd}_{rs}*PIMP_{rd}_{sd}_{rs})/({exp})"

        # Price of imports of good s in region rd from region rs (CES function)
        terms = [f"SHIMPQ_{rd}_{sd}_{rs}**(sigma_bilat_{rd.lower()}_{sd})*(PIMP_{rd}_{sd}_{rs}*icebergcost_{rd.lower()}_{sd}_{rs.lower()})**(1-sigma_bilat_{rd.lower()}_{sd})" for rs in lst_reg if rs != rd]
        exp  = "+".join(terms)
        exp2= f"({exp})**(1/(1-sigma_bilat_{rd.lower()}_{sd}))"
        io.equations[f"PIMP_{rd}_{sd}"] = f"PIMP_{rd}_{sd} := CALPIMP_{rd}_{sd}[{BASEYEAR}]*{exp2}"
        io.identities[f"CALPIMP_{rd}_{sd}"] = f"PIMP_{rd}_{sd}/({exp2})"
        for rs in lst_reg:
            if rs != rd:
                io.equations[f"IMPQ_{rd}_{sd}_{rs}"] = f"IMPQ_{rd}_{sd}_{rs} := if(IMPQ_{rd}_{sd}_{rs}[{BASEYEAR}]<0.001, IMPQ_{rd}_{sd}_{rs}[{BASEYEAR}], CALIMPQ_{rd}_{sd}_{rs}[{BASEYEAR}]*IMPQ_{rd}_{sd}*(((PIMP_{rd}_{sd}_{rs}*icebergcost_{rd.lower()}_{sd}_{rs.lower()})/PIMP_{rd}_{sd}))**(-sigma_bilat_{rd.lower()}_{sd})*SHIMPQ_{rd}_{sd}_{rs}[{BASEYEAR}]**sigma_bilat_{rd.lower()}_{sd})"
                io.identities[f"CALIMPQ_{rd}_{sd}_{rs}"] = f"IMPQ_{rd}_{sd}_{rs}/(IMPQ_{rd}_{sd}*(((PIMP_{rd}_{sd}_{rs}*icebergcost_{rd.lower()}_{sd}_{rs.lower()})/PIMP_{rd}_{sd}))**(-sigma_bilat_{rd.lower()}_{sd})*SHIMPQ_{rd}_{sd}_{rs}[{BASEYEAR}]**sigma_bilat_{rd.lower()}_{sd})"

```

```python
## Values and equations by (r,s) for DOMQ, PADDDTOT
## Equations by (r,s) for IMPQ

for r in lst_reg:
    for ss in lst_sec:
        # Domestic quantity demanded of good s in region r
        io.equations[f"DOMQ_{r}_{ss}"] = f"DOMQ_{r}_{ss} := if(DOMQ_{r}_{ss}[{BASEYEAR}]<0.01,DOMQ_{r}_{ss}[{BASEYEAR}],CALDOMQ_{r}_{ss}[{BASEYEAR}]*ADDDTOTQ_{r}_{ss}*(PPRODHT_{r}_{ss}/PADDDTOT_{r}_{ss})**(-sigma_imp_{r.lower()}_{ss})*SHDOMQ_{r}_{ss}[{BASEYEAR}])"
        io.identities[f"DOMQ_{r}_{ss}"] = f"ADDDTOTQ_{r}_{ss}-IMPQ_{r}_{ss}"

        # Shares of domestic and imported goods in total domestic demand
        io.identities[f"SHDOMQ_{r}_{ss}"] = f"DOMQ_{r}_{ss}*PPRODHT_{r}_{ss}/(DOMQ_{r}_{ss}*PPRODHT_{r}_{ss}+IMPQ_{r}_{ss}*PIMP_{r}_{ss})"
        io.identities[f"SHIMPQ_{r}_{ss}"] = f"IMPQ_{r}_{ss}*PIMP_{r}_{ss}/(DOMQ_{r}_{ss}*PPRODHT_{r}_{ss}+IMPQ_{r}_{ss}*PIMP_{r}_{ss})"
        io.equations[f"SHDOMQ_{r}_{ss}"] = f"SHDOMQ_{r}_{ss} := DOMQ_{r}_{ss}*PPRODHT_{r}_{ss}/(DOMQ_{r}_{ss}*PPRODHT_{r}_{ss}+IMPQ_{r}_{ss}*PIMP_{r}_{ss})"
        io.equations[f"SHIMPQ_{r}_{ss}"] = f"SHIMPQ_{r}_{ss} := IMPQ_{r}_{ss}*PIMP_{r}_{ss}/(DOMQ_{r}_{ss}*PPRODHT_{r}_{ss}+IMPQ_{r}_{ss}*PIMP_{r}_{ss})"

        # Price of total domestic demand for good s in region r (CES function)
        io.identities[f"PADDDTOT_{r}_{ss}"] = f"1"
        
        io.equations[f"PADDDTOT_{r}_{ss}"] = f"PADDDTOT_{r}_{ss} := CALPADDDTOT_{r}_{ss}*(SHDOMQ_{r}_{ss}[{BASEYEAR}]**sigma_imp_{r.lower()}_{ss}*PPRODHT_{r}_{ss}**(1-sigma_imp_{r.lower()}_{ss})+SHIMPQ_{r}_{ss}[{BASEYEAR}]**sigma_imp_{r.lower()}_{ss}*PIMP_{r}_{ss}**(1-sigma_imp_{r.lower()}_{ss}))**(1/(1-sigma_imp_{r.lower()}_{ss}))"
        io.identities[f"CALPADDDTOT_{r}_{ss}"] = f"PADDDTOT_{r}_{ss} /((SHDOMQ_{r}_{ss}[{BASEYEAR}]**sigma_imp_{r.lower()}_{ss}*PPRODHT_{r}_{ss}**(1-sigma_imp_{r.lower()}_{ss})+SHIMPQ_{r}_{ss}[{BASEYEAR}]**sigma_imp_{r.lower()}_{ss}*PIMP_{r}_{ss}**(1-sigma_imp_{r.lower()}_{ss}))**(1/(1-sigma_imp_{r.lower()}_{ss})))"
        io.equations[f"PADDDTOT_{r}_{ss}"] = f"PADDDTOT_{r}_{ss} := CALPADDDTOT_{r}_{ss}*(SHDOMQ_{r}_{ss}[{BASEYEAR}]*PPRODHT_{r}_{ss}**(1-sigma_imp_{r.lower()}_{ss})+SHIMPQ_{r}_{ss}[{BASEYEAR}]*PIMP_{r}_{ss}**(1-sigma_imp_{r.lower()}_{ss}))**(1/(1-sigma_imp_{r.lower()}_{ss}))"
        io.identities[f"CALPADDDTOT_{r}_{ss}"] = f"PADDDTOT_{r}_{ss} /((SHDOMQ_{r}_{ss}[{BASEYEAR}]*PPRODHT_{r}_{ss}**(1-sigma_imp_{r.lower()}_{ss})+SHIMPQ_{r}_{ss}[{BASEYEAR}]*PIMP_{r}_{ss}**(1-sigma_imp_{r.lower()}_{ss}))**(1/(1-sigma_imp_{r.lower()}_{ss})))"


        # Calculation of CALDOMQ (request PADDDTOT)
        io.identities[f"CALDOMQ_{r}_{ss}"] = f"DOMQ_{r}_{ss}/(ADDDTOTQ_{r}_{ss}*(PPRODHT_{r}_{ss}/PADDDTOT_{r}_{ss})**(-sigma_imp_{r.lower()}_{ss})*SHDOMQ_{r}_{ss}[{BASEYEAR}])"


        # Imported quantity demanded of good s in region r
        io.equations[f"IMPQ_{r}_{ss}"] = f"IMPQ_{r}_{ss} := CALIMPQ_{r}_{ss}[{BASEYEAR}]*ADDDTOTQ_{r}_{ss}*(PIMP_{r}_{ss}/PADDDTOT_{r}_{ss})**(-sigma_imp_{r.lower()}_{ss})*SHIMPQ_{r}_{ss}[{BASEYEAR}]"
        io.identities[f"CALIMPQ_{r}_{ss}"] = f"IMPQ_{r}_{ss}/(ADDDTOTQ_{r}_{ss}*(PIMP_{r}_{ss}/PADDDTOT_{r}_{ss})**(-sigma_imp_{r.lower()}_{ss})*SHIMPQ_{r}_{ss}[{BASEYEAR}])"
        
       
        # Domestic production
        io.equations[f"PRODQ_{r}_{ss}"] = f"PRODQ_{r}_{ss} := if(DOMQ_{r}_{ss}[{BASEYEAR}]<0.001,PRODQ_{r}_{ss}[{BASEYEAR}],CALPRODQ_{r}_{ss}*DOMQ_{r}_{ss})"
        io.identities[f"CALPRODQ_{r}_{ss}"] = f"PRODQ_{r}_{ss}/DOMQ_{r}_{ss}"

```

```python
## Values by (rd,sd,rs) for TAXIMP, PIMP, SHIMPQ, CALIMPQ
## Values by (r,s) for PEXPCOMP, CALPEXP, CALPIMP, DOMQ, SHDOMQ, SHIMPQ, PADDDTOT, CALDOMQ, CALIMPQ

lst_idt = ["TAXIMP", "PIMP"]
for item in lst_idt:
    for rd in lst_reg:
        for sd in lst_sec:
            for rs in lst_reg:
                if rs != rd:
                    io.identities.execute(f"{item}_{rd}_{sd}_{rs}", f"{SMPSTRY}", f"{SMPENDY}")
                    

lst_idt = ["PEXPCOMP", "CALPEXP"]
for item in lst_idt:
    for rd in lst_reg:
        for sd in lst_sec:
            io.identities.execute(f"{item}_{rd}_{sd}", f"{SMPSTRY}", f"{SMPENDY}")   

lst_idt = ["SHIMPQ"]
for item in lst_idt:
    for rd in lst_reg:
        for sd in lst_sec:
            for rs in lst_reg:
                if rs != rd:
                    io.identities.execute(f"{item}_{rd}_{sd}_{rs}", f"{SMPSTRY}", f"{SMPENDY}")            

lst_idt = ["CALPIMP"]
for item in lst_idt:
    for rd in lst_reg:
        for sd in lst_sec:
            io.identities.execute(f"{item}_{rd}_{sd}", f"{SMPSTRY}", f"{SMPENDY}")
                      

lst_idt = ["CALIMPQ"]
for item in lst_idt:
    for rd in lst_reg:
        for sd in lst_sec:
            for rs in lst_reg:
                if rs != rd:
                    io.identities.execute(f"{item}_{rd}_{sd}_{rs}", f"{SMPSTRY}", f"{SMPENDY}")                                     


lst_idt = ["DOMQ", "SHDOMQ", "SHIMPQ", "PADDDTOT", "CALPADDDTOT", "CALDOMQ", "CALIMPQ", "CALPRODQ"]
for item in lst_idt:
    for rd in lst_reg:
        for sd in lst_sec:
            io.identities.execute(f"{item}_{rd}_{sd}", f"{SMPSTRY}", f"{SMPENDY}")
```

```python
## Values by (r,skill) for AVEARN
## Values and equations by (r,s,skill) for AVEARN, PEMPHOUR
## Equations by (r,s,skill) for AVEARN, PEMPHOUR

for r in lst_reg:
    for skill in lst_Lab_skill:
        io.identities[f"AVEARN{skill}_{r}"] = f"CMPEMPA{skill}_{r}/DEMPTOT{skill}_{r}"
        io.identities.execute(f"AVEARN{skill}_{r}", f"{SMPSTRY}", f"{SMPENDY}")
        for s in lst_sec:
            io.identities[f"AVEARN{skill}_{r}_{s}"] = f"if(DEMPTOT{skill}_{r}_{s}<0.01 or CMPEMPA{skill}_{r}_{s}<0.01,AVEARN{skill}_{r},CMPEMPA{skill}_{r}_{s}/DEMPTOT{skill}_{r}_{s})"
            io.identities.execute(f"AVEARN{skill}_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")

            io.equations[f"AVEARN{skill}_{r}_{s}"] = f"AVEARN{skill}_{r}_{s} := if(PRODQ_{r}_{s}[{BASEYEAR}]<0.01 or DEMPTOT{skill}_{r}_{s}[{BASEYEAR}]<0.01,AVEARN{skill}_{r}_{s}[{BASEYEAR}],CALAVEARN{skill}_{r}_{s}*(PRODQ_{r}_{s}/DEMPTOT{skill}_{r}_{s})**(av{skill.lower()}_{r.lower()}_pty)*exp((av{skill.lower()}_{r.lower()}_unra)*(UNEMPRA{skill}_{r}-UNEMPRA{skill}B_{r}))*(PCONSH_{r})**av{skill.lower()}_{r.lower()}_pcons*(PCONSH_{r}[-1])**av{skill.lower()}_{r.lower()}_pcons1)"
            io.identities[f"CALAVEARN{skill}_{r}_{s}"] = f"AVEARN{skill}_{r}_{s}/((PRODQ_{r}_{s}/DEMPTOT{skill}_{r}_{s})**(av{skill.lower()}_{r.lower()}_pty)*exp((av{skill.lower()}_{r.lower()}_unra)*(UNEMPRA{skill}_{r}-UNEMPRA{skill}B_{r}))*(PCONSH_{r})**av{skill.lower()}_{r.lower()}_pcons*(PCONSH_{r}[-1])**av{skill.lower()}_{r.lower()}_pcons1)"
            io.identities.execute(f"CALAVEARN{skill}_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")

            io.equations[f"PEMPHOUR{skill}_{r}_{s}"] = f"PEMPHOUR{skill}_{r}_{s} := if(PEMPHOUR{skill}_{r}_{s}[{BASEYEAR}]<0.001,PEMPHOUR{skill}_{r}_{s}[{BASEYEAR}],CALPEMPHOUR{skill}_{r}_{s}*(AVEARN{skill}_{r}_{s}/AVEARN{skill}_{r}_{s}[{BASEYEAR}]))"
            io.identities[f"CALPEMPHOUR{skill}_{r}_{s}"] = f"PEMPHOUR{skill}_{r}_{s}/(AVEARN{skill}_{r}_{s}/AVEARN{skill}_{r}_{s}[{BASEYEAR}])"
            io.identities.execute(f"CALPEMPHOUR{skill}_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")
```

```python
## Equations by (r,skill) for DEMPTOT
## Equations by (r,s,skill) for DEMPTOT

for r in lst_reg:
    for skill in lst_Lab_skill:
        io.equations[f"DEMPTOT{skill}_{r}"] = f"DEMPTOT{skill}_{r} := " + "+".join(f"DEMPTOT{skill}_{r}_{s}" for s in lst_sec)
        for s in lst_sec:
            io.equations[f"DEMPTOT{skill}_{r}_{s}"] = f"DEMPTOT{skill}_{r}_{s} := if(DEMPTOT{skill}_{r}_{s}[{BASEYEAR}]<0.01,DEMPTOT{skill}_{r}_{s}[{BASEYEAR}],CALDEMPTOT{skill}_{r}_{s}*DEMPHOUR{skill}_{r}_{s})"
            io.identities[f"CALDEMPTOT{skill}_{r}_{s}"] = f"DEMPTOT{skill}_{r}_{s}/DEMPHOUR{skill}_{r}_{s}"
            io.identities.execute(f"CALDEMPTOT{skill}_{r}_{s}", f"{SMPSTRY}", f"{SMPENDY}")
```

## Households income

```python
## Values and equations by (r) for GBALPRIMINCH, INCGDISPH, GSAVH, NLENDBORH
## Equations by (r) for CMPEMREH, CMPEMPAH, INCPTYREH, INCPTYPAH, SCNETREH, SCNETPAH, SCOBENOTHREH, SOCBENOTHPAH, SOCBENKIREH, SOCBENKIPAH, 
##                       CURTRANSREH, CURTRANSPAH, ADJPENREH, ADJPENPAH, KTRANSREH, KTRANSPAH, DINVVH, DSTOCKVH, FIXCAPPAH, ACLDNFNPAH, TXDIRPAH

for r in lst_reg:
    for fl in ["RE", "PA"]: # Loop for received and paid
        # In Global Primary Income
        io.equations[f"CMPEM{fl}H_{r}"] = f"CMPEM{fl}H_{r} := CALCMPEM{fl}H_{r}[{BASEYEAR}]*CMPEMPA_{r}"
        io.identities[f"CALCMPEM{fl}H_{r}"] = f"CMPEM{fl}H_{r}/CMPEMPA_{r}"
        # io.equations[f"INCPTY{fl}H_{r}"] = f"INCPTY{fl}H_{r} := CALINCPTY{fl}H_{r}[{BASEYEAR}]*(KSTOCK_{r}*PINV_{r}*RZLR_{r})"
        io.equations[f"INCPTY{fl}H_{r}"] = f"INCPTY{fl}H_{r} := CALINCPTY{fl}H_{r}*(KSTOCK_{r}*PINV_{r}*RZLR_{r})"
        io.identities[f"CALINCPTY{fl}H_{r}"] = f"INCPTY{fl}H_{r}/(KSTOCK_{r}*PINV_{r}*RZLR_{r})"
        # In Gross Available Income
        io.equations[f"SCNET{fl}H_{r}"] = f"SCNET{fl}H_{r} := CALSCNET{fl}H_{r}[{BASEYEAR}]*CMPEMPA_{r}"
        io.identities[f"CALSCNET{fl}H_{r}"] = f"SCNET{fl}H_{r}/CMPEMPA_{r}"
        io.equations[f"SOCBENOTH{fl}H_{r}"] = f"SOCBENOTH{fl}H_{r} := CALSOCBENOTH{fl}H_{r}[{BASEYEAR}]*GDPEXPV_{r}"
        io.identities[f"CALSOCBENOTH{fl}H_{r}"] = f"SOCBENOTH{fl}H_{r}/GDPEXPV_{r}"
        io.equations[f"SOCBENKI{fl}H_{r}"] = f"SOCBENKI{fl}H_{r} := CALSOCBENKI{fl}H_{r}[{BASEYEAR}]*GDPEXPV_{r}"
        io.identities[f"CALSOCBENKI{fl}H_{r}"] = f"SOCBENKI{fl}H_{r}/GDPEXPV_{r}"
        io.equations[f"CURTRANS{fl}H_{r}"] = f"CURTRANS{fl}H_{r} := CALCURTRANS{fl}H_{r}[{BASEYEAR}]*GDPEXPV_{r}"
        io.identities[f"CALCURTRANS{fl}H_{r}"] = f"CURTRANS{fl}H_{r}/GDPEXPV_{r}"
        # In Gross saving
        io.equations[f"ADJPEN{fl}H_{r}"] = f"ADJPEN{fl}H_{r} := CALADJPEN{fl}H_{r}[{BASEYEAR}]*GDPEXPV_{r}"
        io.identities[f"CALADJPEN{fl}H_{r}"] = f"ADJPEN{fl}H_{r}/GDPEXPV_{r}"
        # In Net Lending and Borrwoing
        io.equations[f"KTRANS{fl}H_{r}"] = f"KTRANS{fl}H_{r} := CALKTRANS{fl}H_{r}[{BASEYEAR}]*GDPEXPV_{r}"
        io.identities[f"CALKTRANS{fl}H_{r}"] = f"KTRANS{fl}H_{r}/GDPEXPV_{r}"
        # Execute Idts
        for var in [f"CALCMPEM{fl}H_{r}", f"CALINCPTY{fl}H_{r}", f"CALSCNET{fl}H_{r}", f"CALSOCBENOTH{fl}H_{r}", 
                    f"CALSOCBENKI{fl}H_{r}", f"CALCURTRANS{fl}H_{r}", f"CALADJPEN{fl}H_{r}", f"CALKTRANS{fl}H_{r}"]:
            io.identities.execute(f"{var}", f"{SMPSTRY}", f"{SMPENDY}")
        # Global Primary Income
    io.identities[f"GBALPRIMINCH_{r}"] = f"GOPSUPH_{r}+CMPEMREH_{r}+INCPTYREH_{r}-INCPTYPAH_{r}"
    io.equations[f"GBALPRIMINCH_{r}"] = f"GBALPRIMINCH_{r} := GOPSUPH_{r}+CMPEMREH_{r}+INCPTYREH_{r}-INCPTYPAH_{r}"
    io.identities.execute(f"GBALPRIMINCH_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    
    ## Caution Raw Data Errors, to verify later (probably related to B2G+B3G with B3G missing ??)
    # Correction into GOPSUPH_{r}
    io.variables[f"CORR_{r}"] = f"INCGDISPH_{r}-(GBALPRIMINCH_{r}-TXDIRPAH_{r}+SCNETREH_{r}-SCNETPAH_{r}+SOCBENOTHREH_{r}-SOCBENOTHPAH_{r}+CURTRANSREH_{r}-CURTRANSPAH_{r})"
    io.identities[f"GOPSUPHX_{r}"] = f"GOPSUPH_{r}+CORR_{r}"
    io.identities.execute(f"GOPSUPHX_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    del io.variables[f"CORR_{r}"]
    io.identities[f"GOPSUPH_{r}"] = f"GOPSUPHX_{r}"
    io.identities.execute(f"GOPSUPH_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    del io.variables[f"GOPSUPHX_{r}"]
    io.equations[f"GOPSUPH_{r}"] = f"GOPSUPH_{r} := CALGOPSUPH_{r}[{BASEYEAR}]*GOPSUPV_{r}"
    io.identities[f"CALGOPSUPH_{r}"] = f"GOPSUPH_{r}/GOPSUPV_{r}"
    io.identities.execute(f"CALGOPSUPH_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.identities.execute(f"GBALPRIMINCH_{r}", f"{SMPSTRY}", f"{SMPENDY}")


    # Gross Available Income
    io.equations[f"INCGDISPH_{r}"] = f"INCGDISPH_{r} := GBALPRIMINCH_{r}-TXDIRPAH_{r}+SCNETREH_{r}-SCNETPAH_{r}+SOCBENOTHREH_{r}-SOCBENOTHPAH_{r}+CURTRANSREH_{r}-CURTRANSPAH_{r}"
    io.identities[f"INCGDISPH_{r}"] = f"GBALPRIMINCH_{r}-TXDIRPAH_{r}+SCNETREH_{r}-SCNETPAH_{r}+SOCBENOTHREH_{r}-SOCBENOTHPAH_{r}+CURTRANSREH_{r}-CURTRANSPAH_{r}"
    io.identities.execute(f"INCGDISPH_{r}", f"{SMPSTRY}", f"{SMPENDY}")


    # Gross Saving
    io.identities[f"GSAVH_{r}"] = f"INCGDISPH_{r}-CONSHV_{r}+ADJPENREH_{r}-ADJPENPAH_{r}"
    io.identities.execute(f"GSAVH_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.equations[f"GSAVH_{r}"] = f"GSAVH_{r} := INCGDISPH_{r}-CONSHV_{r}+ADJPENREH_{r}-ADJPENPAH_{r}"
    # Net Lending and Borrowing saving
    io.equations[f"DINVVH_{r}"] = f"DINVVH_{r} := CALDINVVH_{r}*DINVV_{r}"
    io.identities[f"CALDINVVH_{r}"] = f"DINVVH_{r}/DINVV_{r}"
    io.identities.execute(f"CALDINVVH_{r}", f"{SMPSTRY}", f"{SMPENDY}")      
    io.equations[f"DSTOCKVH_{r}"] = f"DSTOCKVH_{r} := CALDSTOCKVH_{r}*DSTOCKV_{r}"
    io.identities[f"CALDSTOCKVH_{r}"] = f"DSTOCKVH_{r}/DSTOCKV_{r}"
    io.identities.execute(f"CALDSTOCKVH_{r}", f"{SMPSTRY}", f"{SMPENDY}")        
    io.equations[f"FIXCAPPAH_{r}"] = f"FIXCAPPAH_{r} := if(DSTOCKV_{r}>-0.01 and DSTOCKV_{r}<0.01,DSTOCKVH_{r}[{BASEYEAR}],CALFIXCAPPAH_{r}*(DINVVH_{r}+DSTOCKVH_{r}))"
    io.identities[f"CALFIXCAPPAH_{r}"] = f"FIXCAPPAH_{r}/(DINVVH_{r}+DSTOCKVH_{r})"
    io.identities.execute(f"CALFIXCAPPAH_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.equations[f"ACLDNFNPAH_{r}"] = f"ACLDNFNPAH_{r} := CALACLDNFNPAH_{r}*GDPEXPV_{r}"
    io.identities[f"CALACLDNFNPAH_{r}"] = f"ACLDNFNPAH_{r}/GDPEXPV_{r}"
    io.identities.execute(f"CALACLDNFNPAH_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.equations[f"NLENDBORH_{r}"] = f"NLENDBORH_{r} := GSAVH_{r}+KTRANSREH_{r}-KTRANSPAH_{r}-FIXCAPPAH_{r}-ACLDNFNPAH_{r}"
    io.identities[f"NLENDBORH_{r}"] = f"GSAVH_{r}+KTRANSREH_{r}-KTRANSPAH_{r}-FIXCAPPAH_{r}-ACLDNFNPAH_{r}"
    io.identities.execute(f"NLENDBORH_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.equations[f"TXDIRPAH_{r}"] = f"TXDIRPAH_{r} := CALTXDIRPAH_{r}*GBALPRIMINCH_{r}"
    io.identities[f"CALTXDIRPAH_{r}"] = f"TXDIRPAH_{r}/GBALPRIMINCH_{r}"
    io.identities.execute(f"CALTXDIRPAH_{r}", f"{SMPSTRY}", f"{SMPENDY}")

    io.equations[f"INCRDISPH_{r}"] = f"INCRDISPH_{r} := INCGDISPH_{r}/PCONSH_{r}"
```

#### Government' accounts

```python
## Values and equations by (r) for GBALPRIMINCG, INCGDISPG, GSAVG, NLENDBORG
## Equations by (r) for CMPEMREH, CMPEMPAH, INCPTYREHG, INCPTYPAG, TXPROIMPREG, SUBTOTPAG, TXDIRREG, TXDIRPAG, CURTRANSREG, CURTRANSPAG, 
##                      ADJPENREG, ADJPENPAG, KTRANSREG, KTRANSPAG, DINVVG, DSTOCKVG, SCNETREG, SOCBENOTHPAG


for r in lst_reg:
    for fl in ["RE", "PA"]: # Loop for received and paid
        # In Global Primary Income
        io.equations[f"INCPTY{fl}G_{r}"] = f"INCPTY{fl}G_{r} := CALINCPTY{fl}G_{r}[{BASEYEAR}]*(KSTOCK_{r}*PINV_{r}*RZLR_{r})"
        io.identities[f"CALINCPTY{fl}G_{r}"] = f"INCPTY{fl}G_{r}/(KSTOCK_{r}*PINV_{r}*RZLR_{r})"
    # Global Primary Income
    io.equations[f"TXPROIMPREG_{r}"] = f"TXPROIMPREG_{r} := CALTXPROIMPREG_{r}*GDPEXPV_{r}"
    io.identities[f"CALTXPROIMPREG_{r}"] = f"TXPROIMPREG_{r}/GDPEXPV_{r}"
    io.identities.execute(f"CALTXPROIMPREG_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.equations[f"SUBTOTPAG_{r}"] = f"SUBTOTPAG_{r} := CALSUBTOTPAG_{r}*GDPEXPV_{r}"
    io.identities[f"CALSUBTOTPAG_{r}"] = f"SUBTOTPAG_{r}/GDPEXPV_{r}"
    io.identities.execute(f"CALSUBTOTPAG_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.identities[f"GBALPRIMINCG_{r}"] = f"GOPSUPG_{r}+TXPROIMPREG_{r}-SUBTOTPAG_{r}+INCPTYREG_{r}-INCPTYPAG_{r}"
    io.equations[f"GBALPRIMINCG_{r}"] = f"GBALPRIMINCG_{r} := GOPSUPG_{r}+TXPROIMPREG_{r}-SUBTOTPAG_{r}+INCPTYREG_{r}-INCPTYPAG_{r}"
    io.identities.execute(f"GBALPRIMINCG_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    for fl in ["RE", "PA"]: # Loop for received and paid
        # # In Gross Available Income
        io.equations[f"TXDIR{fl}G_{r}"] = f"TXDIR{fl}G_{r} := CALTXDIR{fl}G_{r}*GBALPRIMINCG_{r}"
        io.identities[f"CALTXDIR{fl}G_{r}"] = f"TXDIR{fl}G_{r}/GBALPRIMINCG_{r}"
        io.identities.execute(f"CALTXDIR{fl}G_{r}", f"{SMPSTRY}", f"{SMPENDY}")
        io.equations[f"CURTRANS{fl}G_{r}"] = f"CURTRANS{fl}G_{r} := CALCURTRANS{fl}G_{r}[{BASEYEAR}]*GDPEXPV_{r}"
        io.identities[f"CALCURTRANS{fl}G_{r}"] = f"CURTRANS{fl}G_{r}/GDPEXPV_{r}"
        # In Gross saving
        io.equations[f"ADJPEN{fl}G_{r}"] = f"ADJPEN{fl}G_{r} := CALADJPEN{fl}G_{r}[{BASEYEAR}]*GDPEXPV_{r}"
        io.identities[f"CALADJPEN{fl}G_{r}"] = f"ADJPEN{fl}G_{r}/GDPEXPV_{r}"
        # In Net Lending and Borrwoing
        io.equations[f"KTRANS{fl}G_{r}"] = f"KTRANS{fl}G_{r} := CALKTRANS{fl}G_{r}[{BASEYEAR}]*GDPEXPV_{r}"
        io.identities[f"CALKTRANS{fl}G_{r}"] = f"KTRANS{fl}G_{r}/GDPEXPV_{r}"
        # Execute Idts
        for var in [f"CALINCPTY{fl}G_{r}",  f"CALCURTRANS{fl}G_{r}", f"CALADJPEN{fl}G_{r}", f"CALKTRANS{fl}G_{r}"]:
            io.identities.execute(f"{var}", f"{SMPSTRY}", f"{SMPENDY}")

    # Gross Available Income
    io.equations[f"INCGDISPG_{r}"] = f"INCGDISPG_{r} := GBALPRIMINCG_{r}+TXDIRREG_{r}-TXDIRPAG_{r}+SCNETREG_{r}-SOCBENOTHPAG_{r}+CURTRANSREG_{r}-CURTRANSPAG_{r}"
    io.identities[f"INCGDISPG_{r}"] = f"GBALPRIMINCG_{r}+TXDIRREG_{r}-TXDIRPAG_{r}+SCNETREG_{r}-SOCBENOTHPAG_{r}+CURTRANSREG_{r}-CURTRANSPAG_{r}"
    io.identities.execute(f"INCGDISPG_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    # Gross Saving
    io.identities[f"GSAVG_{r}"] = f"INCGDISPG_{r}-CONSGV_{r}+ADJPENREG_{r}-ADJPENPAG_{r}"
    io.identities.execute(f"GSAVG_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.equations[f"GSAVG_{r}"] = f"GSAVG_{r} := INCGDISPG_{r}-CONSGV_{r}+ADJPENREG_{r}-ADJPENPAG_{r}"
    # Net Lending and Borrowing saving
    io.equations[f"DINVVG_{r}"] = f"DINVVG_{r} := CALDINVVG_{r}*DINVV_{r}"
    io.identities[f"CALDINVVG_{r}"] = f"DINVVG_{r}/DINVV_{r}"
    io.identities.execute(f"CALDINVVG_{r}", f"{SMPSTRY}", f"{SMPENDY}")         
    io.equations[f"DSTOCKVG_{r}"] = f"DSTOCKVG_{r} := CALDSTOCKVG_{r}*DSTOCKV_{r}"
    io.identities[f"CALDSTOCKVG_{r}"] = f"DSTOCKVG_{r}/DSTOCKV_{r}"
    io.identities.execute(f"CALDSTOCKVG_{r}", f"{SMPSTRY}", f"{SMPENDY}")        
    io.equations[f"DSTOCKVG_{r}"] = f"DSTOCKVG_{r} := if(DSTOCKV_{r}[{BASEYEAR}]>-0.01 and DSTOCKV_{r}[{BASEYEAR}]<0.01,DSTOCKVG_{r}[{BASEYEAR}],CALDSTOCKVG_{r}*DSTOCKV_{r})"
    io.identities[f"CALFIXCAPPAG_{r}"] = f"FIXCAPPAG_{r}/(DINVVG_{r}+DSTOCKVG_{r})"
    io.identities.execute(f"CALFIXCAPPAG_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.equations[f"ACLDNFNPAG_{r}"] = f"ACLDNFNPAG_{r} := CALACLDNFNPAG_{r}*GDPEXPV_{r}"
    io.identities[f"CALACLDNFNPAG_{r}"] = f"ACLDNFNPAG_{r}/GDPEXPV_{r}"
    io.identities.execute(f"CALACLDNFNPAG_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.identities[f"NLENDBORG_{r}"] = f"GSAVG_{r}+KTRANSREG_{r}-KTRANSPAG_{r}-FIXCAPPAG_{r}-ACLDNFNPAH_{r}"
    io.equations[f"NLENDBORG_{r}"] = f"NLENDBORG_{r} := GSAVG_{r}+KTRANSREG_{r}-KTRANSPAG_{r}-FIXCAPPAG_{r}-ACLDNFNPAH_{r}"
    io.identities.execute(f"NLENDBORG_{r}", f"{SMPSTRY}", f"{SMPENDY}")
    io.equations[f"SCNETREG_{r}"] = f"SCNETREG_{r} := CALSCNETREG_{r}[{BASEYEAR}]*CMPEMPA_{r}"
    io.identities[f"CALSCNETREG_{r}"] = f"SCNETREG_{r}/CMPEMPA_{r}"
    io.equations[f"SOCBENOTHPAG_{r}"] = f"SOCBENOTHPAG_{r} := CALSOCBENOTHPAG_{r}[{BASEYEAR}]*GDPEXPV_{r}"
    io.identities[f"CALSOCBENOTHPAG_{r}"] = f"SOCBENOTHPAG_{r}/GDPEXPV_{r}"
    for var in [f"CALSCNETREG_{r}", f"CALSOCBENOTHPAG_{r}"]:
        io.identities.execute(f"{var}", f"{SMPSTRY}", f"{SMPENDY}")
```

```python
# For Fuel Combustion (FC) and Fugitive Emissions (FE)
lst_ghg = ["CO2", "CH4", "N2O"]
lst_sec_ghg = ["FC", "FE"]
lst_sec_dintenp = ["13", "14", "17", "22", "35", "36", "37", "38", "39"]
lst_sec_dfinenp = [f"{i:02d}" for i in range(1, 60) if f"{i:02d}" not in lst_sec_dintenp]
lst_sec_fe = ["05", "06", "07"]
dict_sec_enr_conso = {"HC": [{"HOIL": "OIL"}, {"HCOM": "COMB"}, {"HGAS": "GAS"}], "TR": [{"IFOI": "OIL"}, {"IFGS": "GAS"}, {"PFOI": "OIL"}]}


dict_sec_ghg = ["FC", "FE"]

for r in lst_reg:
    for ghg in lst_ghg:
        for sec_ghg in lst_sec_ghg:
            if sec_ghg == "FC":
                for sec in lst_sec_dfinenp:
                    io.equations[f"EM{ghg}{sec_ghg}_{r}_{sec}"] = f"EM{ghg}{sec_ghg}_{r}_{sec} := if(EM{ghg}{sec_ghg}_{r}_{sec}[{BASEYEAR}]<0.001,EM{ghg}{sec_ghg}_{r}_{sec}[{BASEYEAR}],CALEM{ghg}{sec_ghg}_{r}_{sec}[{BASEYEAR}]*(em{ghg.lower()}{sec_ghg.lower()}_oil*DFINENP_{r}_{sec}_OIL+em{ghg.lower()}{sec_ghg.lower()}_gas*DFINENP_{r}_{sec}_GAS+em{ghg.lower()}{sec_ghg.lower()}_comb*DFINENP_{r}_{sec}_COMB))"
                    io.identities[f"CALEM{ghg}{sec_ghg}_{r}_{sec}"] = f"EM{ghg}{sec_ghg}_{r}_{sec}/(em{ghg.lower()}{sec_ghg.lower()}_oil*DFINENP_{r}_{sec}_OIL+em{ghg.lower()}{sec_ghg.lower()}_gas*DFINENP_{r}_{sec}_GAS+em{ghg.lower()}{sec_ghg.lower()}_comb*DFINENP_{r}_{sec}_COMB)"
                    io.identities.execute(f"CALEM{ghg}{sec_ghg}_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
                for sec in lst_sec_dintenp:
                    io.equations[f"EM{ghg}{sec_ghg}_{r}_{sec}"] = f"EM{ghg}{sec_ghg}_{r}_{sec} := if(EM{ghg}{sec_ghg}_{r}_{sec}[{BASEYEAR}]<0.001,EM{ghg}{sec_ghg}_{r}_{sec}[{BASEYEAR}],CALEM{ghg}{sec_ghg}_{r}_{sec}[{BASEYEAR}]*(em{ghg.lower()}{sec_ghg.lower()}_oil*(DFINENP_{r}_{sec}_OIL+DINTENP_{r}_{sec}_OIL)+em{ghg.lower()}{sec_ghg.lower()}_gas*(DFINENP_{r}_{sec}_GAS+DINTENP_{r}_{sec}_GAS)+em{ghg.lower()}{sec_ghg.lower()}_comb*(DFINENP_{r}_{sec}_COMB+DINTENP_{r}_{sec}_COMB)))"
                    io.identities[f"CALEM{ghg}{sec_ghg}_{r}_{sec}"] = f"EM{ghg}{sec_ghg}_{r}_{sec}/(em{ghg.lower()}{sec_ghg.lower()}_oil*(DFINENP_{r}_{sec}_OIL+DINTENP_{r}_{sec}_OIL)+em{ghg.lower()}{sec_ghg.lower()}_gas*(DFINENP_{r}_{sec}_GAS+DINTENP_{r}_{sec}_GAS)+em{ghg.lower()}{sec_ghg.lower()}_comb*(DFINENP_{r}_{sec}_COMB+DINTENP_{r}_{sec}_COMB))"
                    io.identities.execute(f"CALEM{ghg}{sec_ghg}_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
            else:
                for sec in lst_sec_fe:
                    io.equations[f"EM{ghg}{sec_ghg}_{r}_{sec}"] = f"EM{ghg}{sec_ghg}_{r}_{sec} := if(EM{ghg}{sec_ghg}_{r}_{sec}[{BASEYEAR}]<0.001,EM{ghg}{sec_ghg}_{r}_{sec}[{BASEYEAR}],CALEM{ghg}{sec_ghg}_{r}_{sec}[{BASEYEAR}]*(em{ghg.lower()}{sec_ghg.lower()}_oil*DFINENP_{r}_{sec}_OIL+em{ghg.lower()}{sec_ghg.lower()}_gas*DFINENP_{r}_{sec}_GAS+em{ghg.lower()}{sec_ghg.lower()}_comb*DFINENP_{r}_{sec}_COMB))"
                    io.identities[f"CALEM{ghg}{sec_ghg}_{r}_{sec}"] = f"EM{ghg}{sec_ghg}_{r}_{sec}/(em{ghg.lower()}{sec_ghg.lower()}_oil*DFINENP_{r}_{sec}_OIL+em{ghg.lower()}{sec_ghg.lower()}_gas*DFINENP_{r}_{sec}_GAS+em{ghg.lower()}{sec_ghg.lower()}_comb*DFINENP_{r}_{sec}_COMB)"
                    io.identities.execute(f"CALEM{ghg}{sec_ghg}_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")

# Households
for r in lst_reg:
    for ghg in lst_ghg:
        for sec_ghg in ["FC"]:
                for sec_conso in ["HC","TR"]:
                    if sec_conso == "HC":
                        io.equations[f"EM{ghg}{sec_ghg}_{r}_{sec_conso}"] = f"EM{ghg}{sec_ghg}_{r}_{sec_conso} := CALEM{ghg}{sec_ghg}_{r}_{sec_conso}[{BASEYEAR}]*(em{ghg.lower()}{sec_ghg.lower()}_oil*DFINENP_{r}_{sec_conso}_OIL+em{ghg.lower()}{sec_ghg.lower()}_gas*DFINENP_{r}_{sec_conso}_GAS+em{ghg.lower()}{sec_ghg.lower()}_comb*DFINENP_{r}_{sec_conso}_COMB)"
                        io.identities[f"CALEM{ghg}{sec_ghg}_{r}_{sec_conso}"] = f"EM{ghg}{sec_ghg}_{r}_{sec_conso}/(em{ghg.lower()}{sec_ghg.lower()}_oil*DFINENP_{r}_{sec_conso}_OIL+em{ghg.lower()}{sec_ghg.lower()}_gas*DFINENP_{r}_{sec_conso}_GAS+em{ghg.lower()}{sec_ghg.lower()}_comb*DFINENP_{r}_{sec_conso}_COMB)"
                        io.identities.execute(f"CALEM{ghg}{sec_ghg}_{r}_{sec_conso}", f"{SMPSTRY}", f"{SMPENDY}")
                    else:
                        io.equations[f"EM{ghg}{sec_ghg}_{r}_{sec_conso}"] = f"EM{ghg}{sec_ghg}_{r}_{sec_conso} := CALEM{ghg}{sec_ghg}_{r}_{sec_conso}[{BASEYEAR}]*(em{ghg.lower()}{sec_ghg.lower()}_oil*DFINENP_{r}_{sec_conso}_OIL+em{ghg.lower()}{sec_ghg.lower()}_gas*DFINENP_{r}_{sec_conso}_GAS)"
                        io.identities[f"CALEM{ghg}{sec_ghg}_{r}_{sec_conso}"] = f"EM{ghg}{sec_ghg}_{r}_{sec_conso}/(em{ghg.lower()}{sec_ghg.lower()}_oil*DFINENP_{r}_{sec_conso}_OIL+em{ghg.lower()}{sec_ghg.lower()}_gas*DFINENP_{r}_{sec_conso}_GAS)"
                        io.identities.execute(f"CALEM{ghg}{sec_ghg}_{r}_{sec_conso}", f"{SMPSTRY}", f"{SMPENDY}")



# For Industrial Emissions (IE), Agriculture (AG), Wastes (WS) and Land use (LU)
dict_sec_ie = {"CO2": ["15","16","17","19","20","21","22","23","24","26","28","30"],
               "CH4": ["15","16","17","19","20","21","22","23","24","26","28","29"],
               "N2O": ["15","16","17","19","20","21","22","23","24","26","28","29"],
               "FGS": ["17","22","23","24","26","28","29"]}
dict_sec_ag = {"CO2": ["01","02"],"CH4": ["01","02"], "N2O":["01","02"]}
dict_sec_ws = {"CO2": ["34"], "CH4": ["34"], "N2O": ["34"]}

# Not including LULUCF in equations
# dict_sec_lu = {"CO2": ["01", "03"], "CH4": ["01", "03"],,"N2O": ["01", "03"],}
#dict_sec_oth = {"IE": dict_sec_ie, "AG": dict_sec_ag, "WS": dict_sec_ws, "LU": dict_sec_lu}

dict_sec_oth = {"IE": dict_sec_ie, "AG": dict_sec_ag, "WS": dict_sec_ws}

for r in lst_reg:
    for em_sec, dict_sec in dict_sec_oth.items():
        for ghg, lst_sec_tmp in dict_sec.items():
            for sec in lst_sec_tmp:
                io.equations[f"EM{ghg}{em_sec}_{r}_{sec}"] = f"EM{ghg}{em_sec}_{r}_{sec} := if(EM{ghg}{em_sec}_{r}_{sec}[{BASEYEAR}]<0.001,EM{ghg}{em_sec}_{r}_{sec}[{BASEYEAR}],CALEM{ghg}{em_sec}_{r}_{sec}*(PRODQ_{r}_{sec}*TPT{ghg}{em_sec}_{r}_{sec}*FACTABAT{ghg}{em_sec}_{r}_{sec}))"
                io.identities[f"CALEM{ghg}{em_sec}_{r}_{sec}"] = f"EM{ghg}{em_sec}_{r}_{sec}/(PRODQ_{r}_{sec}*TPT{ghg}{em_sec}_{r}_{sec}*FACTABAT{ghg}{em_sec}_{r}_{sec})"
                io.equations[f"TPT{ghg}{em_sec}_{r}_{sec}"] =  f"TPT{ghg}{em_sec}_{r}_{sec} := TPT{ghg}{em_sec}_{r}_{sec}[-1]*exp(-PT{ghg}{em_sec}_{r}_{sec})"
                io.identities[f"TPT{ghg}{em_sec}_{r}_{sec}"] = "1"
                io.identities[f"PT{ghg}{em_sec}_{r}_{sec}"] = "0"
                if ghg in ("CO2", "CH4"):
                    io.equations[f"FACTABAT{ghg}{em_sec}_{r}_{sec}"] =  f"FACTABAT{ghg}{em_sec}_{r}_{sec} := 0.75/(0.75+(0.1*({ghg}TAX_{r}*10))**0.5)"
                if ghg == "N2O":
                    io.equations[f"FACTABAT{ghg}{em_sec}_{r}_{sec}"] =  f"FACTABAT{ghg}{em_sec}_{r}_{sec} := 0.9/(0.79+(0.05*({ghg}TAX_{r}*10))**0.75)"
                else:
                    io.equations[f"FACTABAT{ghg}{em_sec}_{r}_{sec}"] =  f"FACTABAT{ghg}{em_sec}_{r}_{sec} := 1/(1+({ghg}TAX_{r}*10))"
                io.identities[f"FACTABAT{ghg}{em_sec}_{r}_{sec}"] = "1"
                io.identities[f"{ghg}TAX_{r}"] = "0"

                io.identities.execute(f"FACTABAT{ghg}{em_sec}_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
                io.identities.execute(f"TPT{ghg}{em_sec}_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
                io.identities.execute(f"PT{ghg}{em_sec}_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
                io.identities.execute(f"{ghg}TAX_{r}", f"{SMPSTRY}", f"{SMPENDY}")
                io.identities.execute(f"CALEM{ghg}{em_sec}_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
```

```python
## Values and equations by (r,src) for LCOE_INV, LCOE_FOAM, LCOE_VOAM, LCOE_FUEL, LCOE_CO2, LCOE

for r in lst_reg:
    for src in lst_pg_src:
        io.identities[f"LCOE_INV_{r}_{src}"] = f"((PGINV_{r}_{src}*PINV_{r}_35/pg_const_{src.lower()})*((1-(1/(1+RZLR_PG_{r}))**(pg_const_{src.lower()}))/(1-(1/(1+RZLR_PG_{r})))))/((365*24*PGCAPFACT_{src}*10**(-3))*(((1-(1/(1+RZLR_PG_{r}))**(pg_const_{src.lower()}+pg_lifet_{src.lower()}))/(1-(1/(1+RZLR_PG_{r}))))-((1-(1/(1+RZLR_PG_{r}))**(pg_const_{src.lower()}))/(1-(1/(1+RZLR_PG_{r}))))))"
        io.identities[f"LCOE_FOAM_{r}_{src}"] = f"(PGFOAM_{r}_{src}/(365*24*PGCAPFACT_{src}*10**(-3)))*(PEMPHOURHS_{r}_35*DEMPHOURHS_{r}_35+PEMPHOURMS_{r}_35*DEMPHOURMS_{r}_35+PEMPHOURLS_{r}_35*DEMPHOURLS_{r}_35)/(PEMPHOURHS_{r}_35[{BASEYEAR}]*DEMPHOURHS_{r}_35+PEMPHOURMS_{r}_35[{BASEYEAR}]*DEMPHOURMS_{r}_35+PEMPHOURLS_{r}_35[{BASEYEAR}]*DEMPHOURLS_{r}_35)"
        io.identities[f"LCOE_VOAM_{r}_{src}"] = f"PGVOAM_{r}_{src}*PMAT_{r}_35"

        if src == "CCCS":
            io.identities[f"LCOE_FUEL_{r}_{src}"] = f"((PENRHT_{r}_35_COMB+TAXENR_{r}_35_COMB)*3.6)*(1/PGENREFF_{r}_35_{src})"
            io.identities[f"LCOE_CO2_{r}_{src}"] = f"((emco2fc_comb*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})*(1-co2ccsrate_{src.lower()})"
        elif src == "GCCS":
            io.identities[f"LCOE_FUEL_{r}_{src}"] = f"((PENRHT_{r}_35_GAS+TAXENR_{r}_35_GAS)*3.6)*(1/PGENREFF_{r}_35_{src})"
            io.identities[f"LCOE_CO2_{r}_{src}"] = f"((emco2fc_gas*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})*(1-co2ccsrate_{src.lower()})"
        elif src == "BCCS":
            io.identities[f"LCOE_FUEL_{r}_{src}"] = f"((PENRHT_{r}_35_SBM+TAXENR_{r}_35_SBM)*3.6)*(1/PGENREFF_{r}_35_{src})"
            io.identities[f"LCOE_CO2_{r}_{src}"] = f"((-1*emco2fc_sbm*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})*co2ccsrate_{src.lower()}"
        elif src in ["COMB", "GAS", "OIL"]:
            io.identities[f"LCOE_FUEL_{r}_{src}"] = f"((PENRHT_{r}_35_{src}+TAXENR_{r}_35_{src})*3.6)*(1/PGENREFF_{r}_35_{src})"
            io.identities[f"LCOE_CO2_{r}_{src}"] = f"((emco2fc_{src.lower()}*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})"
        elif src in ["HYD","GEO","PV","CSP","WDON","WDOF","TID"]:
            io.identities[f"LCOE_FUEL_{r}_{src}"] = "0"
            io.identities[f"LCOE_CO2_{r}_{src}"] = "0"        
        else:
            io.identities[f"LCOE_FUEL_{r}_{src}"] = f"((PENRHT_{r}_35_{src}+TAXENR_{r}_35_{src})*3.6)*(1/PGENREFF_{r}_35_{src})"
            io.identities[f"LCOE_CO2_{r}_{src}"] = "0"
        io.identities[f"LCOE_{r}_{src}"] = f"LCOE_INV_{r}_{src}+LCOE_FOAM_{r}_{src}+LCOE_VOAM_{r}_{src}+LCOE_FUEL_{r}_{src}+LCOE_CO2_{r}_{src}"

        for idt in [f"LCOE_INV_{r}_{src}", f"LCOE_FOAM_{r}_{src}", f"LCOE_VOAM_{r}_{src}", f"LCOE_FUEL_{r}_{src}", f"LCOE_CO2_{r}_{src}", f"LCOE_{r}_{src}"]:
            io.identities.execute(f"{idt}", f"{SMPSTRY}", f"{SMPENDY}")

        io.equations[f"LCOE_INV_{r}_{src}"] = f"LCOE_INV_{r}_{src} := ((PGINV_{r}_{src}*PINV_{r}_35/pg_const_{src.lower()})*((1-(1/(1+RZLR_PG_{r}))**(pg_const_{src.lower()}))/(1-(1/(1+RZLR_PG_{r})))))/((365*24*PGCAPFACT_{src}*10**(-3))*(((1-(1/(1+RZLR_PG_{r}))**(pg_const_{src.lower()}+pg_lifet_{src.lower()}))/(1-(1/(1+RZLR_PG_{r}))))-((1-(1/(1+RZLR_PG_{r}))**(pg_const_{src.lower()}))/(1-(1/(1+RZLR_PG_{r}))))))"
        io.equations[f"LCOE_FOAM_{r}_{src}"] = f"LCOE_FOAM_{r}_{src} := (PGFOAM_{r}_{src}/(365*24*PGCAPFACT_{src}*10**(-3)))*(PEMPHOURHS_{r}_35*DEMPHOURHS_{r}_35+PEMPHOURMS_{r}_35*DEMPHOURMS_{r}_35+PEMPHOURLS_{r}_35*DEMPHOURLS_{r}_35)/(PEMPHOURHS_{r}_35[{BASEYEAR}]*DEMPHOURHS_{r}_35+PEMPHOURMS_{r}_35[{BASEYEAR}]*DEMPHOURMS_{r}_35+PEMPHOURLS_{r}_35[{BASEYEAR}]*DEMPHOURLS_{r}_35)"
        io.equations[f"LCOE_VOAM_{r}_{src}"] = f"LCOE_VOAM_{r}_{src} := PGVOAM_{r}_{src}*PMAT_{r}_35"

        if src == "CCCS":
            io.equations[f"LCOE_FUEL_{r}_{src}"] = f"LCOE_FUEL_{r}_{src} := ((PENRHT_{r}_35_COMB+TAXENR_{r}_35_COMB)*3.6)*(1/PGENREFF_{r}_35_{src})"
            io.equations[f"LCOE_CO2_{r}_{src}"] = f"LCOE_CO2_{r}_{src} := ((emco2fc_comb*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})*(1-co2ccsrate_{src.lower()})"
        elif src == "GCCS":
            io.equations[f"LCOE_FUEL_{r}_{src}"] = f"LCOE_FUEL_{r}_{src} := ((PENRHT_{r}_35_GAS+TAXENR_{r}_35_GAS)*3.6)*(1/PGENREFF_{r}_35_{src})"
            io.equations[f"LCOE_CO2_{r}_{src}"] = f"LCOE_CO2_{r}_{src} := ((emco2fc_gas*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})*(1-co2ccsrate_{src.lower()})"
        elif src == "BCCS":
            io.equations[f"LCOE_FUEL_{r}_{src}"] = f"LCOE_FUEL_{r}_{src} := ((PENRHT_{r}_35_SBM+TAXENR_{r}_35_SBM)*3.6)*(1/PGENREFF_{r}_35_{src})"
            # io.equations[f"LCOE_CO2_{r}_{src}"] = f"LCOE_CO2_{r}_{src} := ((-1*emco2fc_sbm*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})*co2ccsrate_{src.lower()}" # Case with negative cost of CO2
            io.equations[f"LCOE_CO2_{r}_{src}"] = f"LCOE_CO2_{r}_{src} := 0" # case with no payment
            # io.equations[f"LCOE_CO2_{r}_{src}"] = f"LCOE_CO2_{r}_{src} := if(((-1*emco2fc_sbm*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})*co2ccsrate_{src.lower()}<-100,-100,((-1*emco2fc_sbm*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})*co2ccsrate_{src.lower()})"
            
        elif src in ["COMB", "GAS", "OIL"]:
            io.equations[f"LCOE_FUEL_{r}_{src}"] = f"LCOE_FUEL_{r}_{src} := ((PENRHT_{r}_35_{src}+TAXENR_{r}_35_{src})*3.6)*(1/PGENREFF_{r}_35_{src})"
            io.equations[f"LCOE_CO2_{r}_{src}"] = f"LCOE_CO2_{r}_{src} := ((emco2fc_{src.lower()}*CO2TAX_{r}_35*(1/0.2778)))*(1/PGENREFF_{r}_35_{src})"
        elif src in ["HYD","GEO","PV","CSP","WDON","WDOF","TID"]:
            io.equations[f"LCOE_FUEL_{r}_{src}"] = f"LCOE_FUEL_{r}_{src} := 0"
            io.equations[f"LCOE_CO2_{r}_{src}"] = f"LCOE_CO2_{r}_{src} := 0" 
        else:
            io.equations[f"LCOE_FUEL_{r}_{src}"] = f"LCOE_FUEL_{r}_{src} := ((PENRHT_{r}_35_{src}+TAXENR_{r}_35_{src})*3.6)*(1/PGENREFF_{r}_35_{src})"
            io.equations[f"LCOE_CO2_{r}_{src}"] = f"LCOE_CO2_{r}_{src} := 0"
        io.equations[f"LCOE_{r}_{src}"] = f"LCOE_{r}_{src} := LCOE_INV_{r}_{src}+LCOE_FOAM_{r}_{src}+LCOE_VOAM_{r}_{src}+LCOE_FUEL_{r}_{src}+LCOE_CO2_{r}_{src}"


        io.identities[f"LCOE_{r}_{src}"] = f"LCOE_{r}_{src}[{BASEYEAR}]"
        io.identities.execute(f"LCOE_{r}_{src}", f"{SMPSTRY}", "2018Y1")



```

```python
## Values and equations by (r) for PRODP_35_RES, AJELEC, PPRODHT_35
## Values and equations by (r,src) for SHPG, PRODPAL_35, DINTENP_35, DINTENP_36, DINTENP_37, DINTENQ_35, DINTENQ_36, DINTENQ_37

for r in lst_reg:
    exp = [f"PRODP_{r}_35_{src}" for src in lst_src_exo]
    if not exp:
        exp =  ["0"]
    frm = f"PRODP_{r}_35_TOT-(" + "+".join(exp) + ")"
    io.identities[f"PRODP_{r}_35_RES"] = frm
    io.identities.execute(f"PRODP_{r}_35_RES", f"{SMPSTRY}", f"{SMPENDY}")

for r in lst_reg:
    for src in lst_pg_src:
        io.identities[f"SHPG_{r}_{src}"] = f"PRODP_{r}_35_{src}/PRODP_{r}_35_TOT"
        io.identities.execute(f"SHPG_{r}_{src}", f"{SMPSTRY}", f"{SMPENDY}")
        io.equations[f"SHPG_{r}_{src}"] = f"SHPG_{r}_{src} := PRODP_{r}_35_{src}/PRODP_{r}_35_TOT"

for r in lst_reg:
    for src in lst_pg_src:
        io.identities[f"PRODPAL_{r}_35_{src}"] = f"PRODP_{r}_35_{src}"
        io.identities.execute(f"PRODPAL_{r}_35_{src}", f"{SMPSTRY}", f"{SMPENDY}")
    for src in lst_pg_src_noccs:
        io.equations[f"PRODPAL_{r}_35_{src}"] = f"PRODPAL_{r}_35_{src} := if(PRODP_{r}_35_{src}[{BASEYEAR}] < 0.01, PRODP_{r}_35_{src}[{BASEYEAR}],CPRODPAL_{r}_35_{src}*PRODP_{r}_35_RES*(LCOE_{r}_{src}/LCOE_{r}_TOT[-1])**(pglog1_{src.lower()}*(pglog2_{src.lower()}+pglog3_{src.lower()}*exp(pglog4_{src.lower()}*SHPG_{r}_{src}[-1]))))"
        io.identities[f"CPRODPAL_{r}_35_{src}"] = f"PRODPAL_{r}_35_{src}/(PRODP_{r}_35_RES*(LCOE_{r}_{src}/LCOE_{r}_TOT[-1])**(pglog1_{src.lower()}*(pglog2_{src.lower()}+pglog3_{src.lower()}*exp(pglog4_{src.lower()}*SHPG_{r}_{src}[-1]))))"
        io.identities.execute(f"CPRODPAL_{r}_35_{src}", f"{SMPSTRY}", f"{SMPENDY}")
    for src in lst_pg_src_ccs:
        io.variables[f"STR1_{r}_{src}"] = 0
        io.variables[f"STR2_{r}_{src}"] = 0
        io.variables[f"STR3_{r}_{src}"] = 0
        io.equations[f"STR1_{r}_{src}"] = f"STR1_{r}_{src} := if(LCOE_{r}_{src}[-1]<LCOE_{r}_TOT[-1],2,0)"
        io.equations[f"STR2_{r}_{src}"] = f"STR2_{r}_{src} := if(STR1_{r}_{src}>1 and STR1_{r}_{src}[-1]>1,2,0)"
        io.equations[f"STR3_{r}_{src}"] = f"STR3_{r}_{src} := sum(2035Y1, t, STR1_{r}_{src})"
        if src != "BCCS":
            io.equations[f"PRODPAL_{r}_35_{src}"] = f"PRODPAL_{r}_35_{src} := if(t<2035Y1,0.0001,if(STR3_{r}_{src}>1,CPRODPAL_{r}_35_{src}*PRODP_{r}_35_RES*(LCOE_{r}_{src}/LCOE_{r}_TOT[-1])**(pglog1_{src.lower()}*(pglog2_{src.lower()}+pglog3_{src.lower()}*exp(pglog4_{src.lower()}*SHPG_{r}_{src}[-1]))),if(STR3_{r}_{src}>1,CPRODPAL_{r}_35_{src}*PRODP_{r}_35_RES*(LCOE_{r}_{src}/LCOE_{r}_TOT)**(pglog1_{src.lower()}*(pglog2_{src.lower()}+pglog3_{src.lower()}*exp(pglog4_{src.lower()}*SHPG_{r}_{src}[-1]))),0.0001)))"
            io.identities[f"CPRODPAL_{r}_35_{src}"] = f"(PRODP_{r}_35_TOT*0.02)/(PRODP_{r}_35_RES*(LCOE_{r}_{src}/LCOE_{r}_TOT[-1])**(pglog1_{src.lower()}*(pglog2_{src.lower()}+pglog3_{src.lower()}*exp(pglog4_{src.lower()}*SHPG_{r}_{src}[-1]))))"
            io.identities.execute(f"CPRODPAL_{r}_35_{src}", f"{SMPSTRY}", f"{SMPENDY}")  
        else:
            io.identities[f"PRODPAL1_{r}_35_{src}"] = f"PRODPAL_{r}_35_{src}"
            io.identities.execute(f"PRODPAL1_{r}_35_{src}", f"{SMPSTRY}", f"{SMPENDY}")
            io.equations[f"PRODPAL1_{r}_35_{src}"] = f"PRODPAL1_{r}_35_{src} := if(t<2035Y1,0.0001,if(STR3_{r}_{src}>1,CPRODPAL_{r}_35_{src}*PRODP_{r}_35_RES*(LCOE_{r}_{src}/LCOE_{r}_TOT[-1])**(pglog1_{src.lower()}*(pglog2_{src.lower()}+pglog3_{src.lower()}*exp(pglog4_{src.lower()}*SHPG_{r}_{src}[-1]))),if(STR3_{r}_{src}>1,CPRODPAL_{r}_35_{src}*PRODP_{r}_35_RES*(LCOE_{r}_{src}/LCOE_{r}_TOT)**(pglog1_{src.lower()}*(pglog2_{src.lower()}+pglog3_{src.lower()}*exp(pglog4_{src.lower()}*SHPG_{r}_{src}[-1]))),0.0001)))"
            io.identities[f"CPRODPAL_{r}_35_{src}"] = f"(PRODP_{r}_35_TOT*0.02)/(PRODP_{r}_35_RES*(LCOE_{r}_{src}/LCOE_{r}_TOT[-1])**(pglog1_{src.lower()}*(pglog2_{src.lower()}+pglog3_{src.lower()}*exp(pglog4_{src.lower()}*SHPG_{r}_{src}[-1]))))"
            io.identities.execute(f"CPRODPAL_{r}_35_{src}", f"{SMPSTRY}", f"{SMPENDY}")
            io.equations[f"PRODPAL_{r}_35_{src}"] = f"PRODPAL_{r}_35_{src} := if(t<2035Y1,PRODPAL1_{r}_35_{src},if(t<2041Y1,PRODPAL1_{r}_35_{src}*(TIME-2034)/6,PRODPAL1_{r}_35_{src}))"


# DINTENP : Fuel demand in physical unit
for r in lst_reg:
    exp = [f"PRODPAL_{r}_35_{src}" for src in lst_pg_src if src not in lst_src_exo]
    denom = "+".join(exp)
    io.identities[f"AJELEC_{r}"] = f"PRODP_{r}_35_RES/({denom})"
    io.identities.execute(f"AJELEC_{r}")
    io.equations[f"AJELEC_{r}"] = f"AJELEC_{r} := PRODP_{r}_35_RES/({denom})"
    for src in lst_pg_src:
        io.equations[f"PRODP_{r}_35_{src}"] = f"PRODP_{r}_35_{src} := if(PRODPAL_{r}_35_{src}<0.01,PRODP_{r}_35_{src}[-1],PRODPAL_{r}_35_{src}*AJELEC_{r})"
    for src in lst_pg_src_free:    
        for sec in ["35","36","37"]:
            io.identities[f"DINTENP_{r}_{sec}_{src}"] = f"if(PRODP_{r}_35_{src}<0.001,0,PRODP_{r}_35_{src}*(1/PGENREFF_{r}_35_{src})*DINTENQ_{r}_{sec}/(DINTENQ_{r}_35[{BASEYEAR}]+DINTENQ_{r}_36[{BASEYEAR}]+DINTENQ_{r}_37[{BASEYEAR}]))"
            io.identities.execute(f"DINTENP_{r}_{sec}_{src}", f"{SMPSTRY}", f"{SMPENDY}" )
    for src in lst_pg_src_ccs:
        for sec in ["35","36","37"]:
            io.identities[f"DINTENP_{r}_{sec}_{src}"] = f"if(PRODP_{r}_35_{src}<0.001,0,PRODP_{r}_35_{src}*(1/PGENREFF_{r}_35_{src})*DINTENQ_{r}_{sec}/(DINTENQ_{r}_35[{BASEYEAR}]+DINTENQ_{r}_36[{BASEYEAR}]+DINTENQ_{r}_37[{BASEYEAR}]))"
            io.identities.execute(f"DINTENP_{r}_{sec}_{src}", f"{SMPSTRY}", f"{SMPENDY}" )
    for src in lst_pg_src:    
        io.equations[f"DINTENP_{r}_35_{src}"] =  f"DINTENP_{r}_35_{src} := if(PRODP_{r}_35_{src}<0.001,0,PRODP_{r}_35_{src}*(1/PGENREFF_{r}_35_{src})*if((DINTENP_{r}_35_{src}[{BASEYEAR}]+DINTENP_{r}_36_{src}[{BASEYEAR}]+DINTENP_{r}_37_{src}[{BASEYEAR}])<0.01,1,DINTENP_{r}_35_{src}[{BASEYEAR}]/(DINTENP_{r}_35_{src}[{BASEYEAR}]+DINTENP_{r}_36_{src}[{BASEYEAR}]+DINTENP_{r}_37_{src}[{BASEYEAR}])))"
        io.equations[f"DINTENP_{r}_36_{src}"] =  f"DINTENP_{r}_36_{src} := if(PRODP_{r}_35_{src}<0.001,0,PRODP_{r}_35_{src}*(1/PGENREFF_{r}_35_{src})*if((DINTENP_{r}_35_{src}[{BASEYEAR}]+DINTENP_{r}_36_{src}[{BASEYEAR}]+DINTENP_{r}_37_{src}[{BASEYEAR}])<0.01,0,DINTENP_{r}_36_{src}[{BASEYEAR}]/(DINTENP_{r}_35_{src}[{BASEYEAR}]+DINTENP_{r}_36_{src}[{BASEYEAR}]+DINTENP_{r}_37_{src}[{BASEYEAR}])))"
        io.equations[f"DINTENP_{r}_37_{src}"] =  f"DINTENP_{r}_37_{src} := if(PRODP_{r}_35_{src}<0.001,0,PRODP_{r}_35_{src}*(1/PGENREFF_{r}_35_{src})*if((DINTENP_{r}_35_{src}[{BASEYEAR}]+DINTENP_{r}_36_{src}[{BASEYEAR}]+DINTENP_{r}_37_{src}[{BASEYEAR}])<0.01,0,DINTENP_{r}_37_{src}[{BASEYEAR}]/(DINTENP_{r}_35_{src}[{BASEYEAR}]+DINTENP_{r}_36_{src}[{BASEYEAR}]+DINTENP_{r}_37_{src}[{BASEYEAR}])))"


# # DINTENQ : Fuel demand in monetary unit
for r in lst_reg:
    for src in [s for s in lst_pg_src_noccs if s not in lst_pg_src_free]:
        for s in ["35","36","37"]:
            if src == "COMB":
                io.equations[f"DINTENQ_{r}_{s}_{src}"] =  f"DINTENQ_{r}_{s}_{src} := if(DINTENP_{r}_{s}_{src}<0.001,DINTENQ_{r}_{s}_{src}[{BASEYEAR}],CDINTENQ_{r}_{s}_{src}[{BASEYEAR}]*((DINTENP_{r}_{s}_{src}+DINTENP_{r}_{s}_CCCS)/(DINTENP_{r}_{s}_{src}[{BASEYEAR}]+DINTENP_{r}_{s}_CCCS[{BASEYEAR}])))"
                io.identities[f"CDINTENQ_{r}_{s}_{src}"] = f"DINTENQ_{r}_{s}_{src}+DINTENP_{r}_{s}_CCCS"
            if src == "GAS":
                io.equations[f"DINTENQ_{r}_{s}_{src}"] =  f"DINTENQ_{r}_{s}_{src} := if(DINTENP_{r}_{s}_{src}<0.001,DINTENQ_{r}_{s}_{src}[{BASEYEAR}],CDINTENQ_{r}_{s}_{src}[{BASEYEAR}]*((DINTENP_{r}_{s}_{src}+DINTENP_{r}_{s}_GCCS)/(DINTENP_{r}_{s}_{src}[{BASEYEAR}]+DINTENP_{r}_{s}_GCCS[{BASEYEAR}])))"
                io.identities[f"CDINTENQ_{r}_{s}_{src}"] = f"DINTENQ_{r}_{s}_{src}+DINTENP_{r}_{s}_GCCS"
            if src == "SBM":
                io.equations[f"DINTENQ_{r}_{s}_{src}"] =  f"DINTENQ_{r}_{s}_{src} := if(DINTENP_{r}_{s}_{src}<0.001,DINTENQ_{r}_{s}_{src}[{BASEYEAR}],CDINTENQ_{r}_{s}_{src}[{BASEYEAR}]*((DINTENP_{r}_{s}_{src}+DINTENP_{r}_{s}_BCCS)/(DINTENP_{r}_{s}_{src}[{BASEYEAR}]+DINTENP_{r}_{s}_BCCS[{BASEYEAR}])))"
                io.identities[f"CDINTENQ_{r}_{s}_{src}"] = f"DINTENQ_{r}_{s}_{src}+DINTENP_{r}_{s}_BCCS"
            else:
                io.equations[f"DINTENQ_{r}_{s}_{src}"] =  f"DINTENQ_{r}_{s}_{src} := if(DINTENP_{r}_{s}_{src}<0.001,DINTENQ_{r}_{s}_{src}[{BASEYEAR}],CDINTENQ_{r}_{s}_{src}[{BASEYEAR}]*(DINTENP_{r}_{s}_{src}/DINTENP_{r}_{s}_{src}[{BASEYEAR}]))"
                io.identities[f"CDINTENQ_{r}_{s}_{src}"] = f"DINTENQ_{r}_{s}_{src}"
            io.identities.execute(f"CDINTENQ_{r}_{s}_{src}", f"{SMPSTRY}", f"{SMPENDY}")
              

## PPRODHT_35/36/37
for r in lst_reg:
    # for sec in ["35","36","37"]:
    for sec in ["35"]:    
        io.equations[f"PPRODHT_{r}_{sec}"] = f"PPRODHT_{r}_{sec} := CALPPRODHT_{r}_{sec}*LCOE_{r}_TOT"
        io.identities[f"CALPPRODHT_{r}_{sec}"] = f"PPRODHT_{r}_{sec}/LCOE_{r}_TOT"
        io.identities.execute(f"CALPPRODHT_{r}_{sec}", f"{SMPSTRY}", f"{SMPENDY}")
```
