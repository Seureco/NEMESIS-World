```python
# PG Paramters to load
pathpgparam = pathwork  + f"params/"
flpgparam_nm = "Parameters_PG.xlsx"
ld_pgparams = pathpgparam + flpgparam_nm

dict_map_src = {'Nuclear': "NUC", 'Coal': "COMB", 'Coal (CCUS)': "CCCS", 'Gas': "GAS", 'Gas (CCUS)': "GCCS", 'Oil': "OIL",
       'Biomass (wastes)': "IW", 'Biomass (biogas)': "BG", 'Biomass': "SBM", 'Biomass (CCUS)': "BCCS", 'Hydro': "HYD", 'Geothermal': "GEO", 
       'Solar PV': "PV", 'CSP': "SOLT", 'Wind Onshore': "WDON", 'Wind Offshore': "WDOF", 'Tidal': "TID"}

# PGINV
pginv = pd.read_excel(ld_pgparams, sheet_name="Inv_cost",  skiprows=1, header=None)
cols = ["region", "source"] + [f"{y}Y1" for y in range(2019,2051)]
pginv.columns = cols
pginv["source"] = pginv["source"].replace(dict_map_src)
pginv["name"] = "PGINV_" + pginv["region"] + "_" + pginv["source"]
cols_plus = ["name"] + cols
pginv = pginv[cols_plus]
pginv.drop(columns=["region","source"], inplace = True)
pginv = pginv.set_index("name")

io.variables.from_frame(pginv)

# PGFOAM
pgfoam = pd.read_excel(ld_pgparams, sheet_name="FOM_cost",  skiprows=1, header=None)
cols = ["region", "source"] + [f"{y}Y1" for y in range(2019,2051)]
pgfoam.columns = cols
pgfoam["source"] = pgfoam["source"].replace(dict_map_src)
pgfoam["name"] = "PGFOAM_" + pgfoam["region"] + "_" + pgfoam["source"]
cols_plus = ["name"] + cols
pgfoam = pgfoam[cols_plus]
pgfoam.drop(columns=["region","source"], inplace = True)
pgfoam = pgfoam.set_index("name")

io.variables.from_frame(pgfoam)

# PGVOAM
pgvoam = pd.read_excel(ld_pgparams, sheet_name="VOM_cost",  skiprows=1, header=None)
cols = ["region", "source"] + [f"{y}Y1" for y in range(2019,2051)]
pgvoam.columns = cols
pgvoam["source"] = pgvoam["source"].replace(dict_map_src)
pgvoam["name"] = "PGVOAM_" + pgvoam["region"] + "_" + pgvoam["source"]
cols_plus = ["name"] + cols
pgvoam = pgvoam[cols_plus]
pgvoam.drop(columns=["region","source"], inplace = True)
pgvoam = pgvoam.set_index("name")

io.variables.from_frame(pgvoam)


# PGCAPFACT
pgcapf = pd.read_excel(ld_pgparams, sheet_name="CapFact_effective",  skiprows=1, header=None)
cols = ["region", "source"] + [f"{y}Y1" for y in range(2019,2051)]
pgcapf.columns = cols
pgcapf["source"] = pgcapf["source"].replace(dict_map_src)
pgcapf["name"] = "PGCAPFACT_" + pgcapf["source"]
cols_plus = ["name"] + cols
pgcapf= pgcapf[cols_plus]
pgcapf.drop(columns=["region","source"], inplace = True)
pgcapf = pgcapf.set_index("name")

io.variables.from_frame(pgcapf)


# PGENREFFT
pgeff = pd.read_excel(ld_pgparams, sheet_name="NCE",  skiprows=1, header=None)
cols = ["region", "source"] + [f"{y}Y1" for y in range(2019,2051)]
pgeff.columns = cols
pgeff["source"] = pgeff["source"].replace(dict_map_src)
pgeff["name"] = "PGENREFFT_" + pgeff["source"]
cols_plus = ["name"] + cols
pgeff = pgeff[cols_plus]
pgeff.drop(columns=["region","source"], inplace = True)
pgeff = pgeff.set_index("name")

io.variables.from_frame(pgeff)


# pg_const
pglifet = pd.read_excel(ld_pgparams, sheet_name="Lifetime",  skiprows=1, header=None)
cols = ["region", "source", "value"]
pglifet.columns = cols
pglifet["source"] = pglifet["source"].replace(dict_map_src)
pglifet["name"] = "pg_lifet_" + pglifet["source"].str.lower()
cols_plus = ["name"] + cols
pglifet = pglifet[cols_plus]
pglifet.drop(columns=["region","source"], inplace = True)
pglifet = pglifet.set_index("name")

io.scalars.from_frame(pglifet)

# pg_const
pgconst = pd.read_excel(ld_pgparams, sheet_name="ConsTime",  skiprows=1, header=None)
cols = ["region", "source","value"]
pgconst.columns = cols
pgconst["source"] = pgconst["source"].replace(dict_map_src)
pgconst["name"] = "pg_const_" + pgconst["source"].str.lower()
cols_plus = ["name"] + cols
pgconst = pgconst[cols_plus]
pgconst.drop(columns=["region","source"], inplace = True)
pgconst = pgconst.set_index("name")

io.scalars.from_frame(pgconst)


# CCS rates
for css in ["CCCS", "GCCS", "BCCS"]:
    io.scalars[f"co2ccsrate_{css.lower()}"] = 0.9

# Emissions factor for Solid Biomass (in ktCO2/PJ)
io.scalars["emco2fc_sbm"] = 112

```
